## Genetic Algorithm Notes and Observations from the single measurement simulation

**What was noticed:**  

From the GA runs, the **initial population** was not always within the defined bounds.  
Some estimated values were **unrealistic** and generated the exception:

invalid value encountered in sqrt from `turboreactors.py`.  
This same issue sometimes appeared even during:
```python
measured_series = list(h_SCSF(true_traj, S=sensors))
```
meaning some parameter combinations could still produce **non-physical states** ?

---

### What was done
- Implemented the **Genetic Algorithm (GA)** estimation for both 2 and 3 component cases.

- To solve the out-of-bound issue, we forced all outputs to stay **inside the valid bounds** during:

    - **Crossover**
    - **Mutation**
    - **Crossover + Mutation**

This ensures that every individual remains **physically valid** at each generation.



---

### Additional Observation
The current approach gives **good results for early flights**,  
but when using as initial value a point after many flights, the results become **less accurate**.  
From the figure, it seems that the GA **got stuck in a local solution**,   as the population concentrated around a region close to the true value but didn’t fully reach it.

### Next ?
The current approach only consider one measurement but in fact, we should consider:
- More measurements
- The temporal information : The estimation degrade over time
- Update the current GA workflow
- Add early stopping criteria  
- Explore better **initialization strategies**
- Do a crossover ?

Then :
- Add noise in the measurements.


````python
import torch
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from typing import Callable, Tuple, List, Optional


class GeneticAlgorithm:
  """
  Generic, parameterized Genetic Algorithm (GA)
  for solving inverse problems.

  You can customize:
  - Number of genes
  - Population size
  - Mutation rate
  - Crossover rate
  - Tournament size
  - Elitism rate
  - Search bounds
  """

  def __init__(self,
               simulator: Callable,
               target_measurements: List[float],
               sensors: List[str],
               n_genes: int = 2,
               pop_size: int = 100,
               n_generations: int = 50,
               bounds: Optional[List[Tuple[float, float]]] = None,
               mutation_rate: float = 0.1,
               crossover_rate: float = 0.9,
               tournament_size: int = 3,
               n_elite: int = 10,
               mutation_std_ratio: float = 0.05):
    """
    Parameters
    ----------
    simulator : function
        Takes list of genes and returns simulated sensor readings.
    target_measurements : list of float
        Observed target values to match.
    sensors : list of str
        List of sensor names used.
    n_genes : int, default=2
        Number of genes (parameters to optimize).
    pop_size : int, default=100
        Number of individuals per generation.
    n_generations : int, default=50
        Total number of evolution cycles.
    bounds : list of tuple, optional
        [(min, max)] for each gene.
    mutation_rate : float, default=0.1
        Probability of mutating a gene.
    crossover_rate : float, default=0.9
        Probability of performing crossover.
    tournament_size : int, default=3
        Number of individuals in tournament selection.
    n_elite: int, default=10
        Fraction of best individuals to keep unchanged.
    mutation_std_ratio : float, default=0.05
        Fraction of search range used as mutation noise std.
    """
    self.simulator = simulator
    self.target_measurements = np.array(target_measurements)
    self.sensors = sensors
    self.n_sensors = len(sensors)
    self.n_genes = n_genes
    self.pop_size = pop_size
    self.n_generations = n_generations
    self.mutation_rate = mutation_rate
    self.crossover_rate = crossover_rate
    self.tournament_size = tournament_size
    self.n_elite = n_elite
    self.mutation_std_ratio = mutation_std_ratio

    # Default bounds if not given
    if bounds is None:
      self.bounds = [(0.95, 1.02)] * n_genes
    else:
      assert len(bounds) == n_genes, "Bounds list must match number of genes"
      self.bounds = bounds

    # Storage for monitoring
    self.best_fitness_history = []
    self.population_history = []


  # =======================================================
  # --- POPULATION INITIALIZATION ---
  # =======================================================
  # def initialize_population(self) -> torch.Tensor:
  #   """
  #       Create the initial population randomly within given bounds.
  #       Each individual = [η_comp, η_turb].
  #   """
  #   population = torch.zeros((self.pop_size, self.n_genes))
  #   for i, (min_val, max_val) in enumerate(self.bounds):
  #     population[:, i] = torch.rand(self.pop_size) * (max_val - min_val) + min_val
  #   return population

  def initialize_population(self) -> torch.Tensor:
        """
        Create the initial population randomly within given bounds.
        Ensures all individuals remain strictly within [min_val, max_val].
        ach individual = [η_comp, η_turb| or n_comb]
        """
        n_genes = len(self.bounds)
        population = torch.zeros((self.pop_size, n_genes))
    
        for i, (min_val, max_val) in enumerate(self.bounds):
            # Uniform random values
            vals = torch.rand(self.pop_size) * (max_val - min_val) + min_val
    
            # Clamp (safety check)
            vals = torch.clamp(vals, min_val, max_val)
    
            population[:, i] = vals
    
        return population


  # =======================================================
  # --- FITNESS FUNCTION ---
  # =======================================================
  def fitness(self, individual: torch.Tensor) -> float:
    """
    Evaluate one individual (lower error = higher fitness).
    """
    genes = [g.item() for g in individual]
    sim_result = list(self.simulator([genes], S=self.sensors))
      
    simulated_measurements = np.array(sim_result[0]).flatten()

    rel_errors = np.abs(simulated_measurements - self.target_measurements) / np.abs(self.target_measurements)
    total_error = np.mean(rel_errors)

    return -total_error * 1000  # maximize fitness


  def evaluate_population(self, population: torch.Tensor) -> torch.Tensor:
    """Compute fitness for all individuals."""
    return torch.tensor([self.fitness(ind) for ind in population])


  # =======================================================
  # --- SELECTION ---
  # =======================================================
  def selection(self, population: torch.Tensor, fitness_values: torch.Tensor, n_parents: int) -> torch.Tensor:
    """
    Tournament selection of parents.
    """
    parents = torch.zeros((n_parents, self.n_genes))
    for i in range(n_parents):
      candidates_idx = torch.randint(0, self.pop_size, (self.tournament_size,))
      best_idx = candidates_idx[torch.argmax(fitness_values[candidates_idx])]
      parents[i] = population[best_idx]
        
    return parents


  # =======================================================
  # --- CROSSOVER ---
  # =======================================================
  def crossover(self, parent1: torch.Tensor, parent2: torch.Tensor) -> Tuple[torch.Tensor, torch.Tensor]:
    """
    Uniform crossover with random mixing factor α.
    """
    if torch.rand(1).item() > self.crossover_rate:
      return parent1.clone(), parent2.clone()

    alpha = torch.rand(1).item()
    child1 = alpha * parent1 + (1 - alpha) * parent2
    child2 = (1 - alpha) * parent1 + alpha * parent2

    for i, (min_val, max_val) in enumerate(self.bounds):
      child1[i] = torch.clamp(child1[i], min_val, max_val)
      child2[i] = torch.clamp(child2[i], min_val, max_val)

    

    return child1, child2


  # =======================================================
  # --- MUTATION ---
  # =======================================================
  def mutate(self, individual: torch.Tensor) -> torch.Tensor:
    """Random Gaussian mutation applied with probability."""
    mutated = individual.clone()
    for i, (min_val, max_val) in enumerate(self.bounds):
      if torch.rand(1).item() < self.mutation_rate:
        noise_std = (max_val - min_val) * self.mutation_std_ratio
        mutated[i] += torch.normal(0, noise_std, (1,)).item()
        mutated[i] = torch.clamp(mutated[i], min_val, max_val)
          
    return mutated


  # =======================================================
  # --- MAIN EVOLUTION LOOP ---
  # =======================================================
  def evolve(self, verbose_interval: int = 10) -> Tuple[torch.Tensor, List[float]]:
    """
    Full GA evolution loop.
    """
    population = self.initialize_population()

    n_elite = self.n_elite# max(1, int(self.elitism_rate * self.pop_size))

    for generation in range(self.n_generations):
      fitness_values = self.evaluate_population(population)

      # Record best fitness
      best_fitness = torch.max(fitness_values).item()
      self.best_fitness_history.append(best_fitness)
      self.population_history.append(population.clone())

      # Elitism
      elite_idx = torch.argsort(fitness_values, descending=True)[:n_elite]
      elite = population[elite_idx]



      # Selection
      n_parents = self.pop_size - n_elite
      parents = self.selection(population, fitness_values, n_parents)

      # New generation
      new_population = torch.zeros((self.pop_size, self.n_genes))
      new_population[:n_elite] = elite

      # Crossover + mutation
      for i in range(n_elite, self.pop_size, 2):
        parent1 = parents[torch.randint(0, n_parents, (1,)).item()]
        parent2 = parents[torch.randint(0, n_parents, (1,)).item()]
        child1, child2 = self.crossover(parent1, parent2)
        new_population[i] = self.mutate(child1)
        if i + 1 < self.pop_size:
          new_population[i + 1] = self.mutate(child2)

      population = new_population

      # Final global clamp to keep everything in physical range
      for j, (low, high) in enumerate(self.bounds):
        population[:, j] = torch.clamp(population[:, j], low, high)


      # Logging
      if generation % verbose_interval == 0 or generation == self.n_generations - 1:
        best_individual = population[torch.argmax(fitness_values)]
        print(f"Gen {generation:3d} | Best Fitness = {best_fitness:8.3f} | Genes = {best_individual.tolist()}")

      # Print estimated measurement for the last  
      if generation == self.n_generations - 1:
         
          best_individual = population[torch.argmax(fitness_values)]
          # print(f"Gen {generation:3d} | Best Fitness = {best_fitness:8.3f} | Genes = {best_individual.tolist()}")

          sim_result = list(self.simulator([best_individual], S=self.sensors))
      
          simulated_measurements = np.array(sim_result[0]).flatten()

          print (f"simulated measurement : {simulated_measurements}")
      
        

    # Final best
    final_fitness = self.evaluate_population(population)
    best_idx = torch.argmax(final_fitness)
    best_solution = population[best_idx]
    return best_solution, self.best_fitness_history



# =======================================================
# --- PLOTTING UTILITIES ---
# =======================================================
# 2D plot
# def plot_ga_population(ga, best_solution, true_health=None):
#   """Visualize all generations in 2D (works for 2 genes)."""
#   if ga.n_genes != 2:
#     print("--Plot skipped: only supported for 2-gene problems.")
#     return

#   data = []
#   for gen_idx, pop in enumerate(ga.population_history):
#     for x, y in zip(pop[:, 0].numpy(), pop[:, 1].numpy()):
#       data.append({"eta_comp": x, "eta_turb": y, "generation": gen_idx})

#   df = pd.DataFrame(data)
#   plt.figure(figsize=(7, 6))
#   sns.scatterplot(data=df, x="eta_comp", y="eta_turb", hue="generation",
#                   palette="viridis", s=60, alpha=0.8, edgecolor="white")

#   if true_health is not None:
#     plt.scatter(true_health[0], true_health[1], c="lime", s=120, marker="X", label="True")
#   plt.scatter(best_solution[0].item(), best_solution[1].item(),
#               c="red", s=120, marker="*", label="Best")

#   plt.xlabel("η_comp (Compressor)")
#   plt.ylabel("η_turb (Turbine)")
#   plt.title(f"Population Evolution (Sensors: {ga.sensors})")
#   plt.legend()
#   plt.grid(True)
#   plt.show()

import matplotlib.pyplot as plt
import seaborn as sns
from mpl_toolkits.mplot3d import Axes3D

# PLot for 2 and 3 D
def plot_ga_population(ga, best_solution, true_health=None):
    """
    Visualize GA population evolution.
    - 2D scatter if 2 genes (η_comp, η_turb)
    - 3D scatter if 3 genes (η_comp, η_turb, η_comb)
    """

    n_genes = ga.n_genes

    if n_genes not in [2, 3]:
        print(f"--Plot skipped: only supported for 2 or 3 genes (got {n_genes}).")
        return

    # Collect all population data across generations
    data = []
    for gen_idx, pop in enumerate(ga.population_history):
        row = {"generation": gen_idx}
        if n_genes >= 1:
            row["eta_comp"] = pop[:, 0].numpy()
        if n_genes >= 2:
            row["eta_turb"] = pop[:, 1].numpy()
        if n_genes == 3:
            row["eta_comb"] = pop[:, 2].numpy()

        df_gen = pd.DataFrame(row)
        data.append(df_gen)

    df = pd.concat(data, ignore_index=True)

    # ========== 2D case ==========
    if n_genes == 2:
        plt.figure(figsize=(7, 6))
        sns.scatterplot(
            data=df, x="eta_comp", y="eta_turb",
            hue="generation", palette="viridis",
            s=60, alpha=0.8, edgecolor="white"
        )

        if true_health is not None:
            plt.scatter(true_health[0], true_health[1],
                        c="lime", s=120, marker="X", label="True")
        plt.scatter(best_solution[0].item(), best_solution[1].item(),
                    c="red", s=120, marker="*", label="Best")

        plt.xlabel("η_comp (Compressor)")
        plt.ylabel("η_turb (Turbine)")
        plt.title(f"Population Evolution (Sensors: {ga.sensors})")
        plt.legend()
        plt.grid(True)
        plt.show()

    # ========== 3D case ==========
    elif n_genes == 3:
        fig = plt.figure(figsize=(8, 6))
        ax = fig.add_subplot(111, projection='3d')

        sc = ax.scatter(
            df["eta_comp"], df["eta_turb"], df["eta_comb"],
            c=df["generation"], cmap="viridis", s=40, alpha=0.8
        )

        # Add markers for true and best solutions
        if true_health is not None:
            ax.scatter(true_health[0], true_health[1], true_health[2],
                       c="lime", s=100, marker="X", label="True")
        ax.scatter(best_solution[0].item(), best_solution[1].item(), best_solution[2].item(),
                   c="red", s=100, marker="*", label="Best")

        ax.set_xlabel("η_comp (Compressor)")
        ax.set_ylabel("η_turb (Turbine)")
        ax.set_zlabel("η_comb (Combustion)")
        ax.set_title(f"Population Evolution (Sensors: {ga.sensors})")

        plt.legend()
        fig.colorbar(sc, ax=ax, label="Generation")
        plt.show()



def plot_convergence(ga):
  """Plot fitness convergence curve."""
  df = pd.DataFrame({"Generation": range(len(ga.best_fitness_history)),
                     "Best Fitness": ga.best_fitness_history})
  plt.figure(figsize=(7, 4))
  sns.lineplot(data=df, x="Generation", y="Best Fitness", marker="o", linewidth=2.5, color="royalblue")
  plt.grid(True, alpha=0.3)
  plt.title("Fitness Convergence")
  plt.xlabel("Generation")
  plt.ylabel("Best Fitness")
  plt.show()
````

````python
import torch
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from typing import List, Tuple
import time


class TrajectoryEstimator:
    """
    Estimates the full degradation trajectory using Genetic Algorithm
    by solving inverse problems at multiple time steps.
    """
    
    def __init__(self, 
                 ga_class,
                 simulator,
                 sensors: List[str],
                 n_genes: int = 3,
                 **ga_params):
        """
        Parameters
        ----------
        ga_class : class
            The GeneticAlgorithm class to use for each time step
        simulator : callable
            The h_SCSF simulator function
        sensors : list of str
            List of sensor names
        n_genes : int
            Number of genes (health parameters)
        **ga_params : dict
            Additional parameters for GA (pop_size, n_generations, etc.)
        """
        self.ga_class = ga_class
        self.simulator = simulator
        self.sensors = sensors
        self.n_genes = n_genes
        self.ga_params = ga_params
        
        # Storage for results
        self.estimated_trajectory = []
        self.true_trajectory = []
        self.target_measurements_history = []
        self.fitness_histories = []
        self.estimation_errors = []
        self.time_steps = []
        
    def estimate_trajectory(self, 
                           true_traj: np.ndarray,
                           time_indices: List[int] = None,
                           verbose: bool = True,
                           use_adaptive_bounds: bool = True,
                           bounds_margin: float = 0.01) -> np.ndarray:
        """
        Estimate health parameters across multiple time steps.
        
        Parameters
        ----------
        true_traj : np.ndarray
            True health trajectory (time_steps x n_genes)
        time_indices : list of int, optional
            Specific time indices to estimate. If None, uses all.
        verbose : bool
            Whether to print progress
        use_adaptive_bounds : bool, default=True
            If True, uses previous step's solution to narrow search bounds.
            This exploits temporal continuity in degradation.
        bounds_margin : float, default=0.01
            Margin around previous solution for adaptive bounds.
            E.g., if previous η=0.98 and margin=0.01, new bounds=[0.97, 0.99]
            
        Returns
        -------
        estimated_traj : np.ndarray
            Estimated health trajectory
        """
        # Get all sensor measurements
        measured_series = list(self.simulator(true_traj, S=self.sensors))
        
        # Determine which time steps to estimate
        if time_indices is None:
            time_indices = range(len(true_traj))
        
        n_time_steps = len(time_indices)
        
        if verbose:
            print(f"\n{'='*70}")
            print(f"Starting Trajectory Estimation")
            print(f"{'='*70}")
            print(f"Number of time steps: {n_time_steps}")
            print(f"Sensors: {self.sensors}")
            print(f"Health parameters: {self.n_genes}")
            print(f"Adaptive bounds: {use_adaptive_bounds}")
            if use_adaptive_bounds:
                print(f"Bounds margin: ±{bounds_margin}")
            print(f"{'='*70}\n")
        
        # Store original bounds for first iteration
        original_bounds = self.ga_params.get('bounds')
        if original_bounds is None:
            # Default bounds if not specified
            original_bounds = [(0.95, 1.02)] * self.n_genes
        
        previous_solution = None
        
        # Estimate health at each time step
        for step_idx, time_idx in enumerate(time_indices):
            if verbose:
                print(f"\n{'─'*70}")
                print(f"Time Step {step_idx + 1}/{n_time_steps} (Index: {time_idx})")
                print(f"{'─'*70}")
            
            # Get target measurements for this time step
            target_measurements = measured_series[time_idx]
            
            # Adapt bounds based on previous solution
            if use_adaptive_bounds and previous_solution is not None:
                adapted_bounds = []
                for i, prev_val in enumerate(previous_solution):
                    # Create narrow bounds around previous solution
                    lower = max(prev_val - bounds_margin, original_bounds[i][0])
                    upper = min(prev_val + bounds_margin, original_bounds[i][1])
                    adapted_bounds.append((lower, upper))
                
                current_bounds = adapted_bounds
                
                if verbose:
                    print(f"Using adaptive bounds based on previous solution:")
                    for i, (lb, ub) in enumerate(current_bounds):
                        print(f"  Gene {i+1}: [{lb:.4f}, {ub:.4f}] (centered at {previous_solution[i]:.4f})")
            else:
                current_bounds = original_bounds
                if verbose:
                    print(f"Using original bounds: {current_bounds}")
            
            # Create GA instance for this time step with current bounds
            ga_params_copy = self.ga_params.copy()
            ga_params_copy['bounds'] = current_bounds
            
            ga = self.ga_class(
                simulator=self.simulator,
                target_measurements=target_measurements,
                sensors=self.sensors,
                n_genes=self.n_genes,
                **ga_params_copy
            )
            
            # Run evolution
            start_time = time.time()
            best_solution, fitness_history = ga.evolve(
                verbose_interval=self.ga_params.get('n_generations', 50) // 2
            )
            elapsed_time = time.time() - start_time
            
            # Store results
            estimated_health = best_solution.cpu().numpy()
            true_health = true_traj[time_idx]
            
            self.estimated_trajectory.append(estimated_health)
            self.true_trajectory.append(true_health)
            self.target_measurements_history.append(target_measurements)
            self.fitness_histories.append(fitness_history)
            self.time_steps.append(time_idx)
            
            # Calculate errors
            abs_errors = np.abs(estimated_health - true_health)
            rel_errors = abs_errors / np.abs(true_health) * 100
            self.estimation_errors.append(abs_errors)
            
            # Update previous solution for next iteration
            previous_solution = estimated_health
            
            if verbose:
                print(f"\nEstimation Time: {elapsed_time:.2f}s")
                print(f"True Health:      {true_health}")
                print(f"Estimated Health: {estimated_health}")
                print(f"Absolute Errors:  {abs_errors}")
                print(f"Relative Errors:  {rel_errors} %")
        
        if verbose:
            print(f"\n{'='*70}")
            print("Trajectory Estimation Complete!")
            print(f"{'='*70}\n")
        
        return np.array(self.estimated_trajectory)
    
    def plot_trajectory_comparison(self, 
                                   parameter_names: List[str] = None,
                                   figsize: Tuple[int, int] = (14, 10)):
        """
        Plot comparison between true and estimated trajectories.
        """
        if parameter_names is None:
            parameter_names = [f"η_{i+1}" for i in range(self.n_genes)]
        
        true_traj = np.array(self.true_trajectory)
        est_traj = np.array(self.estimated_trajectory)
        time_steps = np.array(self.time_steps)
        
        n_params = len(parameter_names)
        n_cols = 2
        n_rows = (n_params + 1) // 2
        
        fig, axes = plt.subplots(n_rows, n_cols, figsize=figsize)
        axes = axes.flatten() if n_params > 1 else [axes]
        
        for i, (param_name, ax) in enumerate(zip(parameter_names, axes)):
            # Plot true trajectory
            ax.plot(time_steps, true_traj[:, i], 'o-', 
                   label='True', linewidth=2.5, markersize=8, 
                   color='royalblue', alpha=0.7)
            
            # Plot estimated trajectory
            ax.plot(time_steps, est_traj[:, i], 's--', 
                   label='Estimated', linewidth=2, markersize=6,
                   color='crimson', alpha=0.7)
            
            # Styling
            ax.set_xlabel('Time Step', fontsize=11)
            ax.set_ylabel(param_name, fontsize=11)
            ax.set_title(f'{param_name} Evolution', fontsize=12, fontweight='bold')
            ax.legend(loc='best', fontsize=10)
            ax.grid(True, alpha=0.3)
        
        # Remove empty subplots
        for j in range(i + 1, len(axes)):
            fig.delaxes(axes[j])
        
        plt.tight_layout()
        plt.suptitle('Health Parameter Trajectory Comparison', 
                    fontsize=14, fontweight='bold', y=1.02)
        plt.show()
    
    def plot_estimation_errors(self, 
                              parameter_names: List[str] = None,
                              figsize: Tuple[int, int] = (14, 5)):
        """
        Plot absolute and relative errors over time.
        """
        if parameter_names is None:
            parameter_names = [f"η_{i+1}" for i in range(self.n_genes)]
        
        errors = np.array(self.estimation_errors)
        true_traj = np.array(self.true_trajectory)
        time_steps = np.array(self.time_steps)
        
        # Calculate relative errors
        rel_errors = np.abs(errors / true_traj) * 100
        
        fig, axes = plt.subplots(1, 2, figsize=figsize)
        
        # Absolute errors
        for i, param_name in enumerate(parameter_names):
            axes[0].plot(time_steps, errors[:, i], 'o-', 
                        label=param_name, linewidth=2, markersize=6)
        
        axes[0].set_xlabel('Time Step', fontsize=11)
        axes[0].set_ylabel('Absolute Error', fontsize=11)
        axes[0].set_title('Absolute Estimation Errors', fontsize=12, fontweight='bold')
        axes[0].legend(loc='best')
        axes[0].grid(True, alpha=0.3)
        
        # Relative errors (%)
        for i, param_name in enumerate(parameter_names):
            axes[1].plot(time_steps, rel_errors[:, i], 's-', 
                        label=param_name, linewidth=2, markersize=6)
        
        axes[1].set_xlabel('Time Step', fontsize=11)
        axes[1].set_ylabel('Relative Error (%)', fontsize=11)
        axes[1].set_title('Relative Estimation Errors', fontsize=12, fontweight='bold')
        axes[1].legend(loc='best')
        axes[1].grid(True, alpha=0.3)
        
        plt.tight_layout()
        plt.show()
    
    def plot_convergence_heatmap(self, figsize: Tuple[int, int] = (12, 6)):
        """
        Plot fitness convergence as a heatmap across all time steps.
        """
        # Create fitness matrix
        max_gens = max(len(fh) for fh in self.fitness_histories)
        fitness_matrix = np.zeros((len(self.fitness_histories), max_gens))
        
        for i, fh in enumerate(self.fitness_histories):
            fitness_matrix[i, :len(fh)] = fh
            if len(fh) < max_gens:
                fitness_matrix[i, len(fh):] = fh[-1]  # Fill with last value
        
        plt.figure(figsize=figsize)
        sns.heatmap(fitness_matrix, 
                   cmap='YlOrRd',
                   cbar_kws={'label': 'Best Fitness'})
        
        plt.xlabel('Generation', fontsize=11)
        plt.ylabel('Time Step Index', fontsize=11)
        plt.title('Fitness Convergence Across Trajectory', 
                 fontsize=12, fontweight='bold')
        plt.tight_layout()
        plt.show()
    
    def get_summary_statistics(self) -> pd.DataFrame:
        """
        Get summary statistics of estimation performance.
        """
        errors = np.array(self.estimation_errors)
        true_traj = np.array(self.true_trajectory)
        rel_errors = np.abs(errors / true_traj) * 100
        
        stats = {
            'Parameter': [f'η_{i+1}' for i in range(self.n_genes)],
            'Mean Abs Error': np.mean(errors, axis=0),
            # 'Std Abs Error': np.std(errors, axis=0),
            # 'Max Abs Error': np.max(errors, axis=0),
            # 'Mean Rel Error (%)': np.mean(rel_errors, axis=0),
            # 'Std Rel Error (%)': np.std(rel_errors, axis=0),
            # 'Max Rel Error (%)': np.max(rel_errors, axis=0),
        }
        
        return pd.DataFrame(stats)


# =============================================================================
# EXAMPLE USAGE SCRIPT
# =============================================================================

def run_trajectory_estimation_example(GeneticAlgorithm, 
                                      h_SCSF, 
                                      simulate_degradation_trajectory,
                                      time_step_interval: int = 10):
    """
    Complete example of trajectory estimation.
    
    Parameters
    ----------
    GeneticAlgorithm : class
        Your GA class
    h_SCSF : callable
        Simulator function
    simulate_degradation_trajectory : callable
        Function to generate degradation trajectory
    time_step_interval : int, default=10
        Estimate every N-th time step. Use 1 for all steps.
    
    Usage
    ------
    # Estimate every 10th step
    run_trajectory_estimation_example(GeneticAlgorithm, h_SCSF, 
                                     simulate_degradation_trajectory, 
                                     time_step_interval=10)
    
    # Estimate every step
    run_trajectory_estimation_example(GeneticAlgorithm, h_SCSF, 
                                     simulate_degradation_trajectory, 
                                     time_step_interval=1)
    
    # Estimate every 5th step
    run_trajectory_estimation_example(GeneticAlgorithm, h_SCSF, 
                                     simulate_degradation_trajectory, 
                                     time_step_interval=5)
    """
    
    print("="*70)
    print("DEGRADATION TRAJECTORY ESTIMATION WITH GENETIC ALGORITHM")
    print("="*70)
    
    # Step 1: Generate true degradation trajectory
    print("\n[1] Generating true degradation trajectory...")
    true_traj = simulate_degradation_trajectory(
        speed='fast', 
        waterwash='rare', 
        size=5, 
        seed=0
    )
    
    # Configuration
    sensors = ['P3', 'T5', 'Wf']
    parameter_names = ['η_comp', 'η_turb', 'η_comb']
    
    # Select time steps to estimate based on interval
    time_indices = list(range(0, len(true_traj), time_step_interval))
    
    print(f"Total trajectory length: {len(true_traj)}")
    print(f"Time step interval: {time_step_interval}")
    print(f"Estimating at {len(time_indices)} time steps")
    print(f"Time indices: {time_indices[:10]}{'...' if len(time_indices) > 10 else ''}")
    
    # Step 2: Create trajectory estimator
    print("\n[2] Setting up Trajectory Estimator...")
    estimator = TrajectoryEstimator(
        ga_class=GeneticAlgorithm,
        simulator=h_SCSF,
        sensors=sensors,
        n_genes=3,
        pop_size=100,
        n_generations=20,
        bounds=[(0.95, 1.02), (0.95, 1.02), (0.98, 1.02)],
        mutation_rate=0.1,
        crossover_rate=0.9,
        tournament_size=3,
        n_elite=10,
        mutation_std_ratio=0.05
    )
    
    # Step 3: Estimate trajectory
    print("\n[3] Estimating trajectory...")
    overall_start = time.time()
    estimated_traj = estimator.estimate_trajectory(
        true_traj=true_traj,
        time_indices=time_indices,
        verbose=True,
        use_adaptive_bounds=True,  # Use previous solution to narrow search
        bounds_margin=0.01  # ±0.01 around previous solution
    )
    total_time = time.time() - overall_start
    
    print(f"\n{'='*70}")
    print(f"TOTAL EXECUTION TIME: {total_time:.2f}s")
    print(f"Average time per step: {total_time/len(time_indices):.2f}s")
    print(f"{'='*70}\n")
    
    # Step 4: Display summary statistics
    print("\n[4] Summary Statistics:")
    print("-" * 70)
    stats_df = estimator.get_summary_statistics()
    print(stats_df.to_string(index=False))
    
    # Step 5: Visualizations
    print("\n[5] Generating visualizations...")
    
    # Plot trajectory comparison
    estimator.plot_trajectory_comparison(parameter_names=parameter_names)
    
    # Plot estimation errors
    estimator.plot_estimation_errors(parameter_names=parameter_names)
    
    # Plot convergence heatmap
    estimator.plot_convergence_heatmap()
    
    print("\n" + "="*70)
    print("Analysis complete!")
    print("="*70)
    
    return estimator, estimated_traj

````

````python
import torch
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from typing import List, Tuple
import time


class TrajectoryEstimator:
    """
    Estimates the full degradation trajectory using Genetic Algorithm
    by solving inverse problems at multiple time steps.
    """
    
    def __init__(self, 
                 ga_class,
                 simulator,
                 sensors: List[str],
                 n_genes: int = 3,
                 **ga_params):
        """
        Parameters
        ----------
        ga_class : class
            The GeneticAlgorithm class to use for each time step
        simulator : callable
            The h_SCSF simulator function
        sensors : list of str
            List of sensor names
        n_genes : int
            Number of genes (health parameters)
        **ga_params : dict
            Additional parameters for GA (pop_size, n_generations, etc.)
        """
        self.ga_class = ga_class
        self.simulator = simulator
        self.sensors = sensors
        self.n_genes = n_genes
        self.ga_params = ga_params
        
        # Storage for results
        self.estimated_trajectory = []
        self.true_trajectory = []
        self.target_measurements_history = []
        self.fitness_histories = []
        self.estimation_errors = []
        self.time_steps = []
        
    def estimate_trajectory(self, 
                           true_traj: np.ndarray,
                           time_indices: List[int] = None,
                           verbose: bool = True,
                           use_adaptive_bounds: bool = True,
                           bounds_margin: float = 0.01) -> np.ndarray:
        """
        Estimate health parameters across multiple time steps.
        
        Parameters
        ----------
        true_traj : np.ndarray
            True health trajectory (time_steps x n_genes)
        time_indices : list of int, optional
            Specific time indices to estimate. If None, uses all.
        verbose : bool
            Whether to print progress
        use_adaptive_bounds : bool, default=True
            If True, uses previous step's solution to narrow search bounds.
            This exploits temporal continuity in degradation.
        bounds_margin : float, default=0.01
            Margin around previous solution for adaptive bounds.
            E.g., if previous η=0.98 and margin=0.01, new bounds=[0.97, 0.99]
            
        Returns
        -------
        estimated_traj : np.ndarray
            Estimated health trajectory
        """
        # Get all sensor measurements
        measured_series = list(self.simulator(true_traj, S=self.sensors))
        
        # Determine which time steps to estimate
        if time_indices is None:
            time_indices = range(len(true_traj))
        
        n_time_steps = len(time_indices)
        
        if verbose:
            print(f"\n{'='*70}")
            print(f"Starting Trajectory Estimation")
            print(f"{'='*70}")
            print(f"Number of time steps: {n_time_steps}")
            print(f"Sensors: {self.sensors}")
            print(f"Health parameters: {self.n_genes}")
            print(f"Adaptive bounds: {use_adaptive_bounds}")
            if use_adaptive_bounds:
                print(f"Bounds margin: ±{bounds_margin}")
            print(f"{'='*70}\n")
        
        # Store original bounds for first iteration
        original_bounds = self.ga_params.get('bounds', [(0.95, 1.02)] * self.n_genes)
        
        previous_solution = None
        
        # Estimate health at each time step
        for step_idx, time_idx in enumerate(time_indices):
            if verbose:
                print(f"\n{'─'*70}")
                print(f"Time Step {step_idx + 1}/{n_time_steps} (Index: {time_idx})")
                print(f"{'─'*70}")
            
            # Get target measurements for this time step
            target_measurements = measured_series[time_idx]
            
            # Adapt bounds based on previous solution
            if use_adaptive_bounds and previous_solution is not None:
                adapted_bounds = []
                for i, prev_val in enumerate(previous_solution):
                    # Create narrow bounds around previous solution
                    lower = max(prev_val - bounds_margin, original_bounds[i][0])
                    upper = min(prev_val + bounds_margin, original_bounds[i][1])
                    adapted_bounds.append((lower, upper))
                
                current_bounds = adapted_bounds
                
                if verbose:
                    print(f"Using adaptive bounds based on previous solution:")
                    for i, (lb, ub) in enumerate(current_bounds):
                        print(f"  Gene {i+1}: [{lb:.4f}, {ub:.4f}] (centered at {previous_solution[i]:.4f})")
            else:
                current_bounds = original_bounds
                if verbose:
                    print(f"Using original bounds: {current_bounds}")
            
            # Create GA instance for this time step with current bounds
            ga_params_copy = self.ga_params.copy()
            ga_params_copy['bounds'] = current_bounds
            
            ga = self.ga_class(
                simulator=self.simulator,
                target_measurements=target_measurements,
                sensors=self.sensors,
                n_genes=self.n_genes,
                **ga_params_copy
            )
            
            # Run evolution
            start_time = time.time()
            best_solution, fitness_history = ga.evolve(
                verbose_interval=self.ga_params.get('n_generations', 50) // 2
            )
            elapsed_time = time.time() - start_time
            
            # Store results
            estimated_health = best_solution.cpu().numpy()
            true_health = true_traj[time_idx]
            
            self.estimated_trajectory.append(estimated_health)
            self.true_trajectory.append(true_health)
            self.target_measurements_history.append(target_measurements)
            self.fitness_histories.append(fitness_history)
            self.time_steps.append(time_idx)
            
            # Calculate errors
            abs_errors = np.abs(estimated_health - true_health)
            rel_errors = abs_errors / np.abs(true_health) * 100
            self.estimation_errors.append(abs_errors)
            
            # Update previous solution for next iteration
            previous_solution = estimated_health
            
            if verbose:
                print(f"\nEstimation Time: {elapsed_time:.2f}s")
                print(f"True Health:      {true_health}")
                print(f"Estimated Health: {estimated_health}")
                print(f"Absolute Errors:  {abs_errors}")
                print(f"Relative Errors:  {rel_errors} %")
        
        if verbose:
            print(f"\n{'='*70}")
            print("Trajectory Estimation Complete!")
            print(f"{'='*70}\n")
        
        return np.array(self.estimated_trajectory)
    
    def plot_trajectory_comparison(self, 
                                   parameter_names: List[str] = None,
                                   figsize: Tuple[int, int] = (14, 10)):
        """
        Plot comparison between true and estimated trajectories.
        """
        if parameter_names is None:
            parameter_names = [f"η_{i+1}" for i in range(self.n_genes)]
        
        true_traj = np.array(self.true_trajectory)
        est_traj = np.array(self.estimated_trajectory)
        time_steps = np.array(self.time_steps)
        
        n_params = len(parameter_names)
        n_cols = 2
        n_rows = (n_params + 1) // 2
        
        fig, axes = plt.subplots(n_rows, n_cols, figsize=figsize)
        axes = axes.flatten() if n_params > 1 else [axes]
        
        for i, (param_name, ax) in enumerate(zip(parameter_names, axes)):
            # Plot true trajectory
            ax.plot(time_steps, true_traj[:, i], 'o-', 
                   label='True', linewidth=2.5, markersize=8, 
                   color='royalblue', alpha=0.7)
            
            # Plot estimated trajectory
            ax.plot(time_steps, est_traj[:, i], 's--', 
                   label='Estimated', linewidth=2, markersize=6,
                   color='crimson', alpha=0.7)
            
            # Styling
            ax.set_xlabel('Time Step', fontsize=11)
            ax.set_ylabel(param_name, fontsize=11)
            ax.set_title(f'{param_name} Evolution', fontsize=12, fontweight='bold')
            ax.legend(loc='best', fontsize=10)
            ax.grid(True, alpha=0.3)
        
        # Remove empty subplots
        for j in range(i + 1, len(axes)):
            fig.delaxes(axes[j])
        
        plt.tight_layout()
        plt.suptitle('Health Parameter Trajectory Comparison', 
                    fontsize=14, fontweight='bold', y=1.02)
        plt.show()
    
    def plot_estimation_errors(self, 
                              parameter_names: List[str] = None,
                              figsize: Tuple[int, int] = (14, 5)):
        """
        Plot absolute and relative errors over time.
        """
        if parameter_names is None:
            parameter_names = [f"η_{i+1}" for i in range(self.n_genes)]
        
        errors = np.array(self.estimation_errors)
        true_traj = np.array(self.true_trajectory)
        time_steps = np.array(self.time_steps)
        
        # Calculate relative errors
        rel_errors = np.abs(errors / true_traj) * 100
        
        fig, axes = plt.subplots(1, 2, figsize=figsize)
        
        # Absolute errors
        for i, param_name in enumerate(parameter_names):
            axes[0].plot(time_steps, errors[:, i], 'o-', 
                        label=param_name, linewidth=2, markersize=6)
        
        axes[0].set_xlabel('Time Step', fontsize=11)
        axes[0].set_ylabel('Absolute Error', fontsize=11)
        axes[0].set_title('Absolute Estimation Errors', fontsize=12, fontweight='bold')
        axes[0].legend(loc='best')
        axes[0].grid(True, alpha=0.3)
        
        # Relative errors (%)
        for i, param_name in enumerate(parameter_names):
            axes[1].plot(time_steps, rel_errors[:, i], 's-', 
                        label=param_name, linewidth=2, markersize=6)
        
        axes[1].set_xlabel('Time Step', fontsize=11)
        axes[1].set_ylabel('Relative Error (%)', fontsize=11)
        axes[1].set_title('Relative Estimation Errors', fontsize=12, fontweight='bold')
        axes[1].legend(loc='best')
        axes[1].grid(True, alpha=0.3)
        
        plt.tight_layout()
        plt.show()
    
    def plot_convergence_heatmap(self, figsize: Tuple[int, int] = (12, 6)):
        """
        Plot fitness convergence as a heatmap across all time steps.
        """
        # Create fitness matrix
        max_gens = max(len(fh) for fh in self.fitness_histories)
        fitness_matrix = np.zeros((len(self.fitness_histories), max_gens))
        
        for i, fh in enumerate(self.fitness_histories):
            fitness_matrix[i, :len(fh)] = fh
            if len(fh) < max_gens:
                fitness_matrix[i, len(fh):] = fh[-1]  # Fill with last value
        
        plt.figure(figsize=figsize)
        sns.heatmap(fitness_matrix, 
                   cmap='YlOrRd',
                   cbar_kws={'label': 'Best Fitness'})
        
        plt.xlabel('Generation', fontsize=11)
        plt.ylabel('Time Step Index', fontsize=11)
        plt.title('Fitness Convergence Across Trajectory', 
                 fontsize=12, fontweight='bold')
        plt.tight_layout()
        plt.show()
        
    
    def get_summary_statistics(self) -> pd.DataFrame:
        """
        Get summary statistics of estimation performance.
        """
        errors = np.array(self.estimation_errors)
        true_traj = np.array(self.true_trajectory)
        rel_errors = np.abs(errors / true_traj) * 100
        
        stats = {
            'Parameter': [f'η_{i+1}' for i in range(self.n_genes)],
            'Mean Abs Error': np.mean(errors, axis=0),
            'Std Abs Error': np.std(errors, axis=0),
            'Max Abs Error': np.max(errors, axis=0),
            'Mean Rel Error (%)': np.mean(rel_errors, axis=0),
            'Std Rel Error (%)': np.std(rel_errors, axis=0),
            'Max Rel Error (%)': np.max(rel_errors, axis=0),
        }
        
        return pd.DataFrame(stats)


# =============================================================================
# EXAMPLE USAGE SCRIPT
# =============================================================================

def run_trajectory_estimation_example(GeneticAlgorithm, 
                                      h_SCSF, 
                                      simulate_degradation_trajectory):
    """
    Complete example of trajectory estimation.
    
    Usage:
    ------
    run_trajectory_estimation_example(GeneticAlgorithm, h_SCSF, simulate_degradation_trajectory)
    """
    
    print("="*70)
    print("DEGRADATION TRAJECTORY ESTIMATION WITH GENETIC ALGORITHM")
    print("="*70)
    
    # Step 1: Generate true degradation trajectory
    print("\n[1] Generating true degradation trajectory...")
    true_traj = simulate_degradation_trajectory(
        speed='fast', 
        waterwash='rare', 
        size=10, 
        seed=0
    )
    
    # Configuration
    sensors = ['P3', 'T5', 'Wf']
    parameter_names = ['η_comp', 'η_turb', 'η_comb']
    
    # Select time steps to estimate (e.g., every 10th step)
    time_indices = list(range(0, len(true_traj), 10))
    
    print(f"Total trajectory length: {len(true_traj)}")
    print(f"Estimating at {len(time_indices)} time steps")
    print(f"Time indices: {time_indices}")
    
    # Step 2: Create trajectory estimator
    print("\n[2] Setting up Trajectory Estimator...")
    estimator = TrajectoryEstimator(
        ga_class=GeneticAlgorithm,
        simulator=h_SCSF,
        sensors=sensors,
        n_genes=3,
        pop_size=100,
        n_generations=20,
        bounds=[(0.95, 1.02), (0.95, 1.02), (0.98, 1.02)],
        mutation_rate=0.1,
        crossover_rate=0.9,
        tournament_size=3,
        n_elite=10,
        mutation_std_ratio=0.05
    )
    
    # Step 3: Estimate trajectory
    print("\n[3] Estimating trajectory...")
    overall_start = time.time()
    estimated_traj = estimator.estimate_trajectory(
        true_traj=true_traj,
        time_indices=time_indices,
        verbose=True,
        use_adaptive_bounds=True,  # Use previous solution to narrow search
        bounds_margin=0.01  # ±0.01 around previous solution
    )
    total_time = time.time() - overall_start
    
    print(f"\n{'='*70}")
    print(f"TOTAL EXECUTION TIME: {total_time:.2f}s")
    print(f"Average time per step: {total_time/len(time_indices):.2f}s")
    print(f"{'='*70}\n")
    
    # Step 4: Display summary statistics
    print("\n[4] Summary Statistics:")
    print("-" * 70)
    stats_df = estimator.get_summary_statistics()
    print(stats_df.to_string(index=False))
    
    # Step 5: Visualizations
    print("\n[5] Generating visualizations...")
    
    # Plot trajectory comparison
    estimator.plot_trajectory_comparison(parameter_names=parameter_names)
    
    # Plot estimation errors
    estimator.plot_estimation_errors(parameter_names=parameter_names)
    
    # Plot convergence heatmap
    estimator.plot_convergence_heatmap()
    
    print("\n" + "="*70)
    print("Analysis complete!")
    print("="*70)
    
    return estimator, estimated_traj


````

````python
import torch
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from typing import Callable, Tuple, List, Optional
from scipy.stats import qmc


class TemporalGeneticAlgorithm:
    """
    Enhanced Genetic Algorithm for temporal degradation tracking.
    
    Features:
    - Multiple measurements over time
    - Degradation modeling (linear/exponential)
    - Early stopping
    - Advanced initialization strategies
    - Adaptive crossover
    - Noise-robust fitness evaluation
    """

    def __init__(self,
                 simulator: Callable,
                 temporal_measurements: List[List[float]],
                 time_points: List[float],
                 sensors: List[str],
                 n_health_params: int = 2,
                 degradation_model: str = "linear",
                 pop_size: int = 100,
                 n_generations: int = 100,
                 health_bounds: Optional[List[Tuple[float, float]]] = None,
                 rate_bounds: Optional[List[Tuple[float, float]]] = None,
                 mutation_rate: float = 0.1,
                 crossover_rate: float = 0.9,
                 tournament_size: int = 3,
                 n_elite: int = 10,
                 mutation_std_ratio: float = 0.05,
                 init_strategy: str = "hybrid",
                 time_weighting: str = "recent",
                 noise_robust: bool = True):
        """
        Parameters
        ----------
        temporal_measurements : list of lists
            Measurements at each time point, shape [n_timesteps, n_sensors]
        time_points : list of float
            Operating hours for each measurement [0, 1000, 2000, ...]
        n_health_params : int
            Number of health parameters (e.g., 2 for [η_comp, η_turb])
        degradation_model : str
            "linear" or "exponential" degradation
        health_bounds : list of tuple
            Bounds for initial health parameters
        rate_bounds : list of tuple
            Bounds for degradation rates
        init_strategy : str
            "random", "latin_hypercube", "informed", or "hybrid"
        time_weighting : str
            "equal", "recent", or "linear" - how to weight different time points
        noise_robust : bool
            Use median-based error metric for outlier resistance
        """
        self.simulator = simulator
        self.temporal_measurements = np.array(temporal_measurements)
        self.time_points = np.array(time_points)
        self.sensors = sensors
        self.n_sensors = len(sensors)
        self.n_health_params = n_health_params
        self.degradation_model = degradation_model
        self.n_timesteps = len(time_points)
        
        # Total genes = initial health params + degradation rates
        self.n_genes = n_health_params * 2
        
        self.pop_size = pop_size
        self.n_generations = n_generations
        self.mutation_rate = mutation_rate
        self.crossover_rate = crossover_rate
        self.tournament_size = tournament_size
        self.n_elite = n_elite
        self.mutation_std_ratio = mutation_std_ratio
        self.init_strategy = init_strategy
        self.time_weighting = time_weighting
        self.noise_robust = noise_robust

        # Set bounds
        if health_bounds is None:
            self.health_bounds = [(0.85, 1.02)] * n_health_params
        else:
            self.health_bounds = health_bounds
            
        if rate_bounds is None:
            # Default: 0 to 0.00005 per hour (5% degradation over 1000 hours)
            self.rate_bounds = [(0.0, 0.00005)] * n_health_params
        else:
            self.rate_bounds = rate_bounds
        
        # Combined bounds: [health_params, degradation_rates]
        self.bounds = self.health_bounds + self.rate_bounds

        # Compute time weights once
        self.time_weights = self._compute_time_weights()

        # Storage
        self.best_fitness_history = []
        self.mean_fitness_history = []
        self.population_history = []
        self.best_solution_history = []

    def _compute_time_weights(self) -> np.ndarray:
        """Compute weights for different time points."""
        weights = np.ones(self.n_timesteps)
        
        if self.time_weighting == "recent":
            # Exponential decay: recent data more important
            max_time = self.time_points[-1]
            weights = np.exp(-0.0005 * self.time_points / max_time)
        elif self.time_weighting == "linear":
            # Linear increase: later data more important
            weights = np.linspace(0.5, 1.5, self.n_timesteps)
        # else: equal weighting (all 1.0)
        
        return weights / weights.sum()  # Normalize

    # =======================================================
    # --- HEALTH PREDICTION ---
    # =======================================================
    def predict_health_at_time(self, individual: torch.Tensor, time: float) -> List[float]:
        """
        Predict health parameters at given time using degradation model.
        
        individual structure: [η_comp_0, η_turb_0, ..., rate_comp, rate_turb, ...]
        """
        initial_health = individual[:self.n_health_params]
        degradation_rates = individual[self.n_health_params:]
        
        if self.degradation_model == "linear":
            # η(t) = η_0 - rate * t
            current_health = initial_health - degradation_rates * time
        elif self.degradation_model == "exponential":
            # η(t) = η_0 * exp(-rate * t)
            current_health = initial_health * torch.exp(-degradation_rates * time)
        else:
            raise ValueError(f"Unknown degradation model: {self.degradation_model}")
        
        # Clamp to physical bounds
        for i in range(self.n_health_params):
            min_val, max_val = self.health_bounds[i]
            current_health[i] = torch.clamp(current_health[i], min_val, max_val)
        
        return current_health.tolist()

    # =======================================================
    # --- FITNESS FUNCTION ---
    # =======================================================
    def fitness(self, individual: torch.Tensor) -> float:
        """
        Evaluate fitness across ALL time points.
        Lower total error = higher fitness.
        """
        total_error = 0.0
        
        for t_idx, time in enumerate(self.time_points):
            # Predict health at this time
            current_health = self.predict_health_at_time(individual, time)
            
            # Simulate measurements with current health
            try:
                sim_result = list(self.simulator([current_health], S=self.sensors))
                simulated = np.array(sim_result[0]).flatten()
            except Exception as e:
                # Penalize invalid simulations
                return -1e6
            
            # Get target measurements
            target = self.temporal_measurements[t_idx]
            
            # Compute errors
            if self.noise_robust:
                # Median absolute percentage error (robust to outliers)
                rel_errors = np.abs(simulated - target) / (np.abs(target) + 1e-8)
                time_error = np.median(rel_errors)
            else:
                # Mean absolute percentage error
                rel_errors = np.abs(simulated - target) / (np.abs(target) + 1e-8)
                time_error = np.mean(rel_errors)
            
            # Weight by time importance
            total_error += time_error * self.time_weights[t_idx]
        
        return -total_error * 1000  # Maximize fitness

    def evaluate_population(self, population: torch.Tensor) -> torch.Tensor:
        """Compute fitness for all individuals."""
        return torch.tensor([self.fitness(ind) for ind in population])

    # =======================================================
    # --- INITIALIZATION ---
    # =======================================================
    def initialize_population(self) -> torch.Tensor:
        """
        Initialize population using selected strategy.
        """
        population = torch.zeros((self.pop_size, self.n_genes))
        
        if self.init_strategy == "random":
            population = self._init_random()
        elif self.init_strategy == "latin_hypercube":
            population = self._init_latin_hypercube()
        elif self.init_strategy == "informed":
            population = self._init_informed()
        elif self.init_strategy == "hybrid":
            # 40% Latin hypercube, 40% informed, 20% random
            n_lhs = int(0.4 * self.pop_size)
            n_informed = int(0.4 * self.pop_size)
            n_random = self.pop_size - n_lhs - n_informed
            
            population[:n_lhs] = self._init_latin_hypercube()[:n_lhs]
            population[n_lhs:n_lhs+n_informed] = self._init_informed()[:n_informed]
            population[n_lhs+n_informed:] = self._init_random()[:n_random]
        else:
            raise ValueError(f"Unknown init strategy: {self.init_strategy}")
        
        return population

    def _init_random(self) -> torch.Tensor:
        """Uniform random initialization."""
        population = torch.zeros((self.pop_size, self.n_genes))
        for i, (min_val, max_val) in enumerate(self.bounds):
            population[:, i] = torch.rand(self.pop_size) * (max_val - min_val) + min_val
        return population

    def _init_latin_hypercube(self) -> torch.Tensor:
        """Latin Hypercube Sampling for better space coverage."""
        sampler = qmc.LatinHypercube(d=self.n_genes)
        samples = sampler.random(n=self.pop_size)
        
        population = torch.zeros((self.pop_size, self.n_genes))
        for i, (min_val, max_val) in enumerate(self.bounds):
            population[:, i] = torch.tensor(samples[:, i] * (max_val - min_val) + min_val)
        
        return population

    def _init_informed(self) -> torch.Tensor:
        """Domain-informed initialization."""
        population = torch.zeros((self.pop_size, self.n_genes))
        
        for i in range(self.pop_size):
            # Initial health: bias towards healthy (95-100%)
            for j in range(self.n_health_params):
                min_val, max_val = self.health_bounds[j]
                population[i, j] = torch.rand(1) * 0.05 * (max_val - min_val) + 0.95 * max_val
            
            # Degradation rates: small positive values
            for j in range(self.n_health_params):
                min_val, max_val = self.rate_bounds[j]
                population[i, self.n_health_params + j] = torch.rand(1) * (max_val - min_val) * 0.5 + min_val
        
        return population

    # =======================================================
    # --- SELECTION ---
    # =======================================================
    def selection(self, population: torch.Tensor, fitness_values: torch.Tensor, n_parents: int) -> torch.Tensor:
        """Tournament selection."""
        parents = torch.zeros((n_parents, self.n_genes))
        for i in range(n_parents):
            candidates_idx = torch.randint(0, self.pop_size, (self.tournament_size,))
            best_idx = candidates_idx[torch.argmax(fitness_values[candidates_idx])]
            parents[i] = population[best_idx]
        return parents

    # =======================================================
    # --- CROSSOVER ---
    # =======================================================
    def adaptive_crossover(self, parent1: torch.Tensor, parent2: torch.Tensor, generation: int) -> Tuple[torch.Tensor, torch.Tensor]:
        """
        Adaptive crossover that adjusts based on generation progress.
        """
        if torch.rand(1).item() > self.crossover_rate:
            return parent1.clone(), parent2.clone()

        progress = generation / self.n_generations
        
        if progress < 0.5:
            # Early generations: exploration (uniform)
            alpha = torch.rand(1).item()
        else:
            # Later generations: exploitation (favor better parent)
            fitness1 = self.fitness(parent1)
            fitness2 = self.fitness(parent2)
            alpha = 0.7 if fitness1 > fitness2 else 0.3

        child1 = alpha * parent1 + (1 - alpha) * parent2
        child2 = (1 - alpha) * parent1 + alpha * parent2

        # Clamp to bounds
        for i, (min_val, max_val) in enumerate(self.bounds):
            child1[i] = torch.clamp(child1[i], min_val, max_val)
            child2[i] = torch.clamp(child2[i], min_val, max_val)

        return child1, child2

    # =======================================================
    # --- MUTATION ---
    # =======================================================
    def mutate(self, individual: torch.Tensor) -> torch.Tensor:
        """Gaussian mutation with adaptive step size."""
        mutated = individual.clone()
        for i, (min_val, max_val) in enumerate(self.bounds):
            if torch.rand(1).item() < self.mutation_rate:
                noise_std = (max_val - min_val) * self.mutation_std_ratio
                mutated[i] += torch.normal(0, noise_std, (1,)).item()
                mutated[i] = torch.clamp(mutated[i], min_val, max_val)
        return mutated

    # =======================================================
    # --- EVOLUTION LOOP ---
    # =======================================================
    def evolve(self, 
               verbose_interval: int = 10,
               patience: int = 30,
               min_improvement: float = 1e-4) -> Tuple[torch.Tensor, dict]:
        """
        Main evolution loop with early stopping.
        
        Returns
        -------
        best_solution : torch.Tensor
        history : dict with convergence metrics
        """
        population = self.initialize_population()
        
        best_fitness_ever = -np.inf
        generations_without_improvement = 0
        
        print(f"\n{'='*70}")
        print(f"Starting Temporal GA Evolution")
        print(f"{'='*70}")
        print(f"Time points: {self.time_points.tolist()}")
        print(f"Degradation model: {self.degradation_model}")
        print(f"Initialization: {self.init_strategy}")
        print(f"{'='*70}\n")

        for generation in range(self.n_generations):
            fitness_values = self.evaluate_population(population)
            
            best_fitness = torch.max(fitness_values).item()
            mean_fitness = torch.mean(fitness_values).item()
            best_idx = torch.argmax(fitness_values)
            best_individual = population[best_idx]
            
            # Record history
            self.best_fitness_history.append(best_fitness)
            self.mean_fitness_history.append(mean_fitness)
            self.population_history.append(population.clone())
            self.best_solution_history.append(best_individual.clone())

            # Early stopping check
            if best_fitness - best_fitness_ever > min_improvement:
                best_fitness_ever = best_fitness
                generations_without_improvement = 0
            else:
                generations_without_improvement += 1

            if generations_without_improvement >= patience:
                print(f"\n{'='*70}")
                print(f"Early stopping at generation {generation}")
                print(f"No improvement for {patience} generations")
                print(f"{'='*70}\n")
                break

            # Elitism
            elite_idx = torch.argsort(fitness_values, descending=True)[:self.n_elite]
            elite = population[elite_idx]

            # Selection
            n_parents = self.pop_size - self.n_elite
            parents = self.selection(population, fitness_values, n_parents)

            # Create new generation
            new_population = torch.zeros((self.pop_size, self.n_genes))
            new_population[:self.n_elite] = elite

            # Crossover + Mutation
            for i in range(self.n_elite, self.pop_size, 2):
                parent1 = parents[torch.randint(0, n_parents, (1,)).item()]
                parent2 = parents[torch.randint(0, n_parents, (1,)).item()]
                child1, child2 = self.adaptive_crossover(parent1, parent2, generation)
                new_population[i] = self.mutate(child1)
                if i + 1 < self.pop_size:
                    new_population[i + 1] = self.mutate(child2)

            population = new_population

            # Logging
            if generation % verbose_interval == 0 or generation == self.n_generations - 1:
                print(f"Gen {generation:4d} | Best: {best_fitness:8.2f} | Mean: {mean_fitness:8.2f} | Genes: {best_individual.tolist()}")

        # Final evaluation
        print(f"\n{'='*70}")
        print("Evolution Complete - Final Results")
        print(f"{'='*70}")
        
        final_fitness = self.evaluate_population(population)
        best_idx = torch.argmax(final_fitness)
        best_solution = population[best_idx]
        
        # Display predicted degradation trajectory
        print(f"\nBest Solution:")
        print(f"  Initial Health: {best_solution[:self.n_health_params].tolist()}")
        print(f"  Degradation Rates: {best_solution[self.n_health_params:].tolist()}")
        
        print(f"\nPredicted Health Trajectory:")
        for t_idx, time in enumerate(self.time_points):
            health = self.predict_health_at_time(best_solution, time)
            print(f"  t={time:6.0f}h: {health}")
            
            # Show measurement comparison
            sim_result = list(self.simulator([health], S=self.sensors))
            simulated = np.array(sim_result[0]).flatten()
            target = self.temporal_measurements[t_idx]
            errors = np.abs(simulated - target) / np.abs(target) * 100
            print(f"           Simulated: {simulated}")
            print(f"           Target:    {target}")
            print(f"           Error (%): {errors}\n")
        
        history = {
            'best_fitness': self.best_fitness_history,
            'mean_fitness': self.mean_fitness_history,
            'population': self.population_history,
            'best_solutions': self.best_solution_history
        }
        
        return best_solution, history


# =======================================================
# --- UTILITY FUNCTIONS ---
# =======================================================
def add_measurement_noise(measurements: np.ndarray, 
                         noise_level: float = 0.02,
                         outlier_prob: float = 0.05) -> np.ndarray:
    """
    Add realistic sensor noise to measurements.
    
    Parameters
    ----------
    measurements : array
        Clean measurements
    noise_level : float
        Gaussian noise standard deviation (relative)
    outlier_prob : float
        Probability of outlier for each measurement
    """
    noisy = measurements.copy()
    
    # Gaussian noise
    noise = np.random.normal(0, noise_level, measurements.shape)
    noisy = noisy * (1 + noise)
    
    # Occasional outliers
    outlier_mask = np.random.rand(*measurements.shape) < outlier_prob
    if outlier_mask.any():
        outliers = np.random.uniform(-0.1, 0.1, outlier_mask.sum())
        noisy[outlier_mask] *= (1 + outliers)
    
    return noisy


# =======================================================
# --- PLOTTING ---
# =======================================================
def plot_convergence(ga, figsize=(12, 5)):
    """Plot fitness convergence curves."""
    fig, axes = plt.subplots(1, 2, figsize=figsize)
    
    # Best fitness
    axes[0].plot(ga.best_fitness_history, linewidth=2.5, color='royalblue', label='Best')
    axes[0].plot(ga.mean_fitness_history, linewidth=2, color='coral', alpha=0.7, label='Mean')
    axes[0].set_xlabel('Generation', fontsize=11)
    axes[0].set_ylabel('Fitness', fontsize=11)
    axes[0].set_title('Fitness Convergence', fontsize=12, fontweight='bold')
    axes[0].grid(True, alpha=0.3)
    axes[0].legend()
    
    # Fitness improvement rate
    improvement = np.diff(ga.best_fitness_history)
    axes[1].plot(improvement, linewidth=2, color='green')
    axes[1].axhline(y=0, color='red', linestyle='--', alpha=0.5)
    axes[1].set_xlabel('Generation', fontsize=11)
    axes[1].set_ylabel('Fitness Improvement', fontsize=11)
    axes[1].set_title('Improvement Rate', fontsize=12, fontweight='bold')
    axes[1].grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()


def plot_degradation_trajectory(ga, best_solution, true_trajectory=None, figsize=(12, 6)):
    """
    Plot predicted vs actual degradation over time.
    """
    n_params = ga.n_health_params
    param_names = ['η_comp', 'η_turb', 'η_comb'][:n_params]
    
    fig, axes = plt.subplots(1, n_params, figsize=figsize)
    if n_params == 1:
        axes = [axes]
    
    for i in range(n_params):
        # Predicted trajectory
        predicted = []
        for time in ga.time_points:
            health = ga.predict_health_at_time(best_solution, time)
            predicted.append(health[i])
        
        axes[i].plot(ga.time_points, predicted, 'o-', linewidth=2.5, 
                    markersize=8, label='Predicted', color='royalblue')
        
        # True trajectory if available
        if true_trajectory is not None:
            true_vals = [true_trajectory(t)[i] for t in ga.time_points]
            axes[i].plot(ga.time_points, true_vals, 's--', linewidth=2, 
                        markersize=7, label='True', color='red', alpha=0.7)
        
        axes[i].set_xlabel('Operating Hours', fontsize=11)
        axes[i].set_ylabel(param_names[i], fontsize=11)
        axes[i].set_title(f'{param_names[i]} Degradation', fontsize=12, fontweight='bold')
        axes[i].grid(True, alpha=0.3)
        axes[i].legend()
    
    plt.tight_layout()
    plt.show()


def plot_measurement_fit(ga, best_solution, figsize=(14, 4)):
    """
    Plot simulated vs actual measurements at each time point.
    """
    n_times = len(ga.time_points)
    fig, axes = plt.subplots(1, n_times, figsize=figsize, sharey=True)
    if n_times == 1:
        axes = [axes]
    
    for t_idx, time in enumerate(ga.time_points):
        # Get simulated and actual measurements
        health = ga.predict_health_at_time(best_solution, time)
        sim_result = list(ga.simulator([health], S=ga.sensors))
        simulated = np.array(sim_result[0]).flatten()
        target = ga.temporal_measurements[t_idx]
        
        x = np.arange(len(ga.sensors))
        width = 0.35
        
        axes[t_idx].bar(x - width/2, target, width, label='Measured', color='coral', alpha=0.8)
        axes[t_idx].bar(x + width/2, simulated, width, label='Simulated', color='royalblue', alpha=0.8)
        axes[t_idx].set_xlabel('Sensors', fontsize=10)
        if t_idx == 0:
            axes[t_idx].set_ylabel('Value', fontsize=11)
        axes[t_idx].set_title(f't = {time:.0f}h', fontsize=11, fontweight='bold')
        axes[t_idx].set_xticks(x)
        axes[t_idx].set_xticklabels(ga.sensors, rotation=45, ha='right')
        axes[t_idx].legend(fontsize=9)
        axes[t_idx].grid(True, alpha=0.3, axis='y')
    
    plt.tight_layout()
    plt.show()
````

````python
# Generate temporal measurements (with degradation)
time_points = [0, 500, 1000, 1500, 2000]  # operating hours
temporal_measurements = [
    [1.45, 2.30, 0.85],  # measurements at t=0
    [1.42, 2.25, 0.83],  # t=500
    [1.39, 2.20, 0.81],  # t=1000
    # ... etc
]

# Add noise
noisy_measurements = add_measurement_noise(np.array(temporal_measurements), 
                                           noise_level=0.02)

# Create GA
ga = TemporalGeneticAlgorithm(
    simulator= h_SCSF,
    temporal_measurements=noisy_measurements,
    time_points=time_points,
    sensors=['P3', 'T3', 'Wf'],
    n_health_params=2,  # η_comp, η_turb
    degradation_model="linear",
    pop_size=150,
    n_generations=100,
    init_strategy="hybrid",
    time_weighting="recent",
    noise_robust=True
)

# Evolve
best_solution, history = ga.evolve(verbose_interval=10, patience=30)

# Visualize
plot_convergence(ga)
plot_degradation_trajectory(ga, best_solution)
plot_measurement_fit(ga, best_solution)
````

````python
import torch
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from typing import Callable, Tuple, List, Optional
from scipy.stats import qmc


class TemporalGeneticAlgorithm:
    """
    Enhanced Genetic Algorithm for temporal degradation tracking.
    
    Features:
    - Multiple measurements over time
    - Degradation modeling (linear/exponential)
    - Early stopping
    - Advanced initialization strategies
    - Adaptive crossover
    - Noise-robust fitness evaluation
    """

    def __init__(self,
                 simulator: Callable,
                 temporal_measurements: List[List[float]],
                 time_points: List[float],
                 sensors: List[str],
                 n_health_params: int = 2,
                 degradation_model: str = "linear",
                 pop_size: int = 100,
                 n_generations: int = 100,
                 health_bounds: Optional[List[Tuple[float, float]]] = None,
                 rate_bounds: Optional[List[Tuple[float, float]]] = None,
                 mutation_rate: float = 0.1,
                 crossover_rate: float = 0.9,
                 tournament_size: int = 3,
                 n_elite: int = 10,
                 mutation_std_ratio: float = 0.05,
                 init_strategy: str = "hybrid",
                 time_weighting: str = "recent",
                 noise_robust: bool = True):
        """
        Parameters
        ----------
        temporal_measurements : list of lists
            Measurements at each time point, shape [n_timesteps, n_sensors]
        time_points : list of float
            Operating hours for each measurement [0, 1000, 2000, ...]
        n_health_params : int
            Number of health parameters (e.g., 2 for [η_comp, η_turb])
        degradation_model : str
            "linear" or "exponential" degradation
        health_bounds : list of tuple
            Bounds for initial health parameters
        rate_bounds : list of tuple
            Bounds for degradation rates
        init_strategy : str
            "random", "latin_hypercube", "informed", or "hybrid"
        time_weighting : str
            "equal", "recent", or "linear" - how to weight different time points
        noise_robust : bool
            Use median-based error metric for outlier resistance
        """
        self.simulator = simulator
        self.temporal_measurements = np.array(temporal_measurements)
        self.time_points = np.array(time_points)
        self.sensors = sensors
        self.n_sensors = len(sensors)
        self.n_health_params = n_health_params
        self.degradation_model = degradation_model
        self.n_timesteps = len(time_points)
        
        # Total genes = initial health params + degradation rates
        self.n_genes = n_health_params * 2
        
        self.pop_size = pop_size
        self.n_generations = n_generations
        self.mutation_rate = mutation_rate
        self.crossover_rate = crossover_rate
        self.tournament_size = tournament_size
        self.n_elite = n_elite
        self.mutation_std_ratio = mutation_std_ratio
        self.init_strategy = init_strategy
        self.time_weighting = time_weighting
        self.noise_robust = noise_robust

        # Set bounds
        if health_bounds is None:
            self.health_bounds = [(0.85, 1.02)] * n_health_params
        else:
            self.health_bounds = health_bounds
            
        if rate_bounds is None:
            # Default: 0 to 0.00005 per hour (5% degradation over 1000 hours)
            self.rate_bounds = [(0.0, 0.00005)] * n_health_params
        else:
            self.rate_bounds = rate_bounds
        
        # Combined bounds: [health_params, degradation_rates]
        self.bounds = self.health_bounds + self.rate_bounds

        # Compute time weights once
        self.time_weights = self._compute_time_weights()

        # Storage
        self.best_fitness_history = []
        self.mean_fitness_history = []
        self.population_history = []
        self.best_solution_history = []

    def _compute_time_weights(self) -> np.ndarray:
        """Compute weights for different time points."""
        weights = np.ones(self.n_timesteps)
        
        if self.time_weighting == "recent":
            # Exponential decay: recent data more important
            max_time = self.time_points[-1]
            weights = np.exp(-0.0005 * self.time_points / max_time)
        elif self.time_weighting == "linear":
            # Linear increase: later data more important
            weights = np.linspace(0.5, 1.5, self.n_timesteps)
        # else: equal weighting (all 1.0)
        
        return weights / weights.sum()  # Normalize

    # =======================================================
    # --- HEALTH PREDICTION ---
    # =======================================================
    def predict_health_at_time(self, individual: torch.Tensor, time: float) -> List[float]:
        """
        Predict health parameters at given time using degradation model.
        
        individual structure: [η_comp_0, η_turb_0, ..., rate_comp, rate_turb, ...]
        """
        initial_health = individual[:self.n_health_params]
        degradation_rates = individual[self.n_health_params:]
        
        if self.degradation_model == "linear":
            # η(t) = η_0 - rate * t
            current_health = initial_health - degradation_rates * time
        elif self.degradation_model == "exponential":
            # η(t) = η_0 * exp(-rate * t)
            current_health = initial_health * torch.exp(-degradation_rates * time)
        else:
            raise ValueError(f"Unknown degradation model: {self.degradation_model}")
        
        # Clamp to physical bounds
        for i in range(self.n_health_params):
            min_val, max_val = self.health_bounds[i]
            current_health[i] = torch.clamp(current_health[i], min_val, max_val)
        
        return current_health.tolist()

    # =======================================================
    # --- FITNESS FUNCTION ---
    # =======================================================
    def fitness(self, individual: torch.Tensor) -> float:
        """
        Evaluate fitness across ALL time points.
        Lower total error = higher fitness.
        """
        total_error = 0.0
        
        for t_idx, time in enumerate(self.time_points):
            # Predict health at this time
            current_health = self.predict_health_at_time(individual, time)
            
            # Simulate measurements with current health
            try:
                sim_result = list(self.simulator([current_health], S=self.sensors))
                simulated = np.array(sim_result[0]).flatten()
            except Exception as e:
                # Penalize invalid simulations
                return -1e6
            
            # Get target measurements
            target = self.temporal_measurements[t_idx]
            
            # Compute errors
            if self.noise_robust:
                # Median absolute percentage error (robust to outliers)
                rel_errors = np.abs(simulated - target) / (np.abs(target) + 1e-8)
                time_error = np.median(rel_errors)
            else:
                # Mean absolute percentage error
                rel_errors = np.abs(simulated - target) / (np.abs(target) + 1e-8)
                time_error = np.mean(rel_errors)
            
            # Weight by time importance
            total_error += time_error * self.time_weights[t_idx]
        
        return -total_error * 1000  # Maximize fitness

    def evaluate_population(self, population: torch.Tensor) -> torch.Tensor:
        """Compute fitness for all individuals."""
        return torch.tensor([self.fitness(ind) for ind in population])

    # =======================================================
    # --- INITIALIZATION ---
    # =======================================================
    def initialize_population(self) -> torch.Tensor:
        """
        Initialize population using selected strategy.
        """
        population = torch.zeros((self.pop_size, self.n_genes))
        
        if self.init_strategy == "random":
            population = self._init_random()
        elif self.init_strategy == "latin_hypercube":
            population = self._init_latin_hypercube()
        elif self.init_strategy == "informed":
            population = self._init_informed()
        elif self.init_strategy == "hybrid":
            # 40% Latin hypercube, 40% informed, 20% random
            n_lhs = int(0.4 * self.pop_size)
            n_informed = int(0.4 * self.pop_size)
            n_random = self.pop_size - n_lhs - n_informed
            
            population[:n_lhs] = self._init_latin_hypercube()[:n_lhs]
            population[n_lhs:n_lhs+n_informed] = self._init_informed()[:n_informed]
            population[n_lhs+n_informed:] = self._init_random()[:n_random]
        else:
            raise ValueError(f"Unknown init strategy: {self.init_strategy}")
        
        return population

    def _init_random(self) -> torch.Tensor:
        """Uniform random initialization."""
        population = torch.zeros((self.pop_size, self.n_genes))
        for i, (min_val, max_val) in enumerate(self.bounds):
            population[:, i] = torch.rand(self.pop_size) * (max_val - min_val) + min_val
        return population

    def _init_latin_hypercube(self) -> torch.Tensor:
        """Latin Hypercube Sampling for better space coverage."""
        sampler = qmc.LatinHypercube(d=self.n_genes)
        samples = sampler.random(n=self.pop_size)
        
        population = torch.zeros((self.pop_size, self.n_genes))
        for i, (min_val, max_val) in enumerate(self.bounds):
            population[:, i] = torch.tensor(samples[:, i] * (max_val - min_val) + min_val)
        
        return population

    def _init_informed(self) -> torch.Tensor:
        """Domain-informed initialization."""
        population = torch.zeros((self.pop_size, self.n_genes))
        
        for i in range(self.pop_size):
            # Initial health: bias towards healthy (95-100%)
            for j in range(self.n_health_params):
                min_val, max_val = self.health_bounds[j]
                population[i, j] = torch.rand(1) * 0.05 * (max_val - min_val) + 0.95 * max_val
            
            # Degradation rates: small positive values
            for j in range(self.n_health_params):
                min_val, max_val = self.rate_bounds[j]
                population[i, self.n_health_params + j] = torch.rand(1) * (max_val - min_val) * 0.5 + min_val
        
        return population

    # =======================================================
    # --- SELECTION ---
    # =======================================================
    def selection(self, population: torch.Tensor, fitness_values: torch.Tensor, n_parents: int) -> torch.Tensor:
        """Tournament selection."""
        parents = torch.zeros((n_parents, self.n_genes))
        for i in range(n_parents):
            candidates_idx = torch.randint(0, self.pop_size, (self.tournament_size,))
            best_idx = candidates_idx[torch.argmax(fitness_values[candidates_idx])]
            parents[i] = population[best_idx]
        return parents

    # =======================================================
    # --- CROSSOVER ---
    # =======================================================
    def adaptive_crossover(self, parent1: torch.Tensor, parent2: torch.Tensor, generation: int) -> Tuple[torch.Tensor, torch.Tensor]:
        """
        Adaptive crossover that adjusts based on generation progress.
        """
        if torch.rand(1).item() > self.crossover_rate:
            return parent1.clone(), parent2.clone()

        progress = generation / self.n_generations
        
        if progress < 0.5:
            # Early generations: exploration (uniform)
            alpha = torch.rand(1).item()
        else:
            # Later generations: exploitation (favor better parent)
            fitness1 = self.fitness(parent1)
            fitness2 = self.fitness(parent2)
            alpha = 0.7 if fitness1 > fitness2 else 0.3

        child1 = alpha * parent1 + (1 - alpha) * parent2
        child2 = (1 - alpha) * parent1 + alpha * parent2

        # Clamp to bounds
        for i, (min_val, max_val) in enumerate(self.bounds):
            child1[i] = torch.clamp(child1[i], min_val, max_val)
            child2[i] = torch.clamp(child2[i], min_val, max_val)

        return child1, child2

    # =======================================================
    # --- MUTATION ---
    # =======================================================
    def mutate(self, individual: torch.Tensor) -> torch.Tensor:
        """Gaussian mutation with adaptive step size."""
        mutated = individual.clone()
        for i, (min_val, max_val) in enumerate(self.bounds):
            if torch.rand(1).item() < self.mutation_rate:
                noise_std = (max_val - min_val) * self.mutation_std_ratio
                mutated[i] += torch.normal(0, noise_std, (1,)).item()
                mutated[i] = torch.clamp(mutated[i], min_val, max_val)
        return mutated

    # =======================================================
    # --- EVOLUTION LOOP ---
    # =======================================================
    def evolve(self, 
               verbose_interval: int = 10,
               patience: int = 30,
               min_improvement: float = 1e-4) -> Tuple[torch.Tensor, dict]:
        """
        Main evolution loop with early stopping.
        
        Returns
        -------
        best_solution : torch.Tensor
        history : dict with convergence metrics
        """
        population = self.initialize_population()
        
        best_fitness_ever = -np.inf
        generations_without_improvement = 0
        
        print(f"\n{'='*70}")
        print(f"Starting Temporal GA Evolution")
        print(f"{'='*70}")
        print(f"Time points: {self.time_points.tolist()}")
        print(f"Degradation model: {self.degradation_model}")
        print(f"Initialization: {self.init_strategy}")
        print(f"{'='*70}\n")

        for generation in range(self.n_generations):
            fitness_values = self.evaluate_population(population)
            
            best_fitness = torch.max(fitness_values).item()
            mean_fitness = torch.mean(fitness_values).item()
            best_idx = torch.argmax(fitness_values)
            best_individual = population[best_idx]
            
            # Record history
            self.best_fitness_history.append(best_fitness)
            self.mean_fitness_history.append(mean_fitness)
            self.population_history.append(population.clone())
            self.best_solution_history.append(best_individual.clone())

            # Early stopping check
            if best_fitness - best_fitness_ever > min_improvement:
                best_fitness_ever = best_fitness
                generations_without_improvement = 0
            else:
                generations_without_improvement += 1

            if generations_without_improvement >= patience:
                print(f"\n{'='*70}")
                print(f"Early stopping at generation {generation}")
                print(f"No improvement for {patience} generations")
                print(f"{'='*70}\n")
                break

            # Elitism
            elite_idx = torch.argsort(fitness_values, descending=True)[:self.n_elite]
            elite = population[elite_idx]

            # Selection
            n_parents = self.pop_size - self.n_elite
            parents = self.selection(population, fitness_values, n_parents)

            # Create new generation
            new_population = torch.zeros((self.pop_size, self.n_genes))
            new_population[:self.n_elite] = elite

            # Crossover + Mutation
            for i in range(self.n_elite, self.pop_size, 2):
                parent1 = parents[torch.randint(0, n_parents, (1,)).item()]
                parent2 = parents[torch.randint(0, n_parents, (1,)).item()]
                child1, child2 = self.adaptive_crossover(parent1, parent2, generation)
                new_population[i] = self.mutate(child1)
                if i + 1 < self.pop_size:
                    new_population[i + 1] = self.mutate(child2)

            population = new_population

            # Logging
            if generation % verbose_interval == 0 or generation == self.n_generations - 1:
                print(f"Gen {generation:4d} | Best: {best_fitness:8.2f} | Mean: {mean_fitness:8.2f} | Genes: {best_individual.tolist()}")

        # Final evaluation
        print(f"\n{'='*70}")
        print("Evolution Complete - Final Results")
        print(f"{'='*70}")
        
        final_fitness = self.evaluate_population(population)
        best_idx = torch.argmax(final_fitness)
        best_solution = population[best_idx]
        
        # Display predicted degradation trajectory
        print(f"\nBest Solution:")
        print(f"  Initial Health: {best_solution[:self.n_health_params].tolist()}")
        print(f"  Degradation Rates: {best_solution[self.n_health_params:].tolist()}")
        
        print(f"\nPredicted Health Trajectory:")
        for t_idx, time in enumerate(self.time_points):
            health = self.predict_health_at_time(best_solution, time)
            print(f"  t={time:6.0f}h: {health}")
            
            # Show measurement comparison
            sim_result = list(self.simulator([health], S=self.sensors))
            simulated = np.array(sim_result[0]).flatten()
            target = self.temporal_measurements[t_idx]
            errors = np.abs(simulated - target) / np.abs(target) * 100
            print(f"           Simulated: {simulated}")
            print(f"           Target:    {target}")
            print(f"           Error (%): {errors}\n")
        
        history = {
            'best_fitness': self.best_fitness_history,
            'mean_fitness': self.mean_fitness_history,
            'population': self.population_history,
            'best_solutions': self.best_solution_history
        }
        
        return best_solution, history


# =======================================================
# --- UTILITY FUNCTIONS ---
# =======================================================
def add_measurement_noise(measurements: np.ndarray, 
                         noise_level: float = 0.02,
                         outlier_prob: float = 0.05) -> np.ndarray:
    """
    Add realistic sensor noise to measurements.
    
    Parameters
    ----------
    measurements : array
        Clean measurements
    noise_level : float
        Gaussian noise standard deviation (relative)
    outlier_prob : float
        Probability of outlier for each measurement
    """
    noisy = measurements.copy()
    
    # Gaussian noise
    noise = np.random.normal(0, noise_level, measurements.shape)
    noisy = noisy * (1 + noise)
    
    # Occasional outliers
    outlier_mask = np.random.rand(*measurements.shape) < outlier_prob
    if outlier_mask.any():
        outliers = np.random.uniform(-0.1, 0.1, outlier_mask.sum())
        noisy[outlier_mask] *= (1 + outliers)
    
    return noisy


def prepare_temporal_data(true_trajectory, simulator, sensors, time_indices, noise_level=0.0):
    """
    Helper function to prepare temporal measurements from a degradation trajectory.
    
    Parameters
    ----------
    true_trajectory : array-like
        Full degradation trajectory, shape [n_samples, n_health_params]
    simulator : callable
        Simulator function
    sensors : list of str
        Sensor names
    time_indices : list of int
        Indices to sample from trajectory (e.g., [0, 10, 20, 30])
    noise_level : float
        Noise level to add (0.0 = no noise)
    
    Returns
    -------
    temporal_measurements : list of lists
        Measurements at each time point
    time_points : list of float
        Corresponding time points (in hours, assuming 1 sample = 1 hour)
    sampled_health : array
        True health states at sampled time points
    """
    temporal_measurements = []
    time_points = []
    sampled_health = []
    
    for idx in time_indices:
        # Get true health at this time
        health = true_trajectory[idx]
        sampled_health.append(health)
        
        # Simulate measurements
        sim_result = list(simulator([health], S=sensors))
        measurements = np.array(sim_result[0]).flatten()
        
        # Add noise if requested
        if noise_level > 0:
            measurements = add_measurement_noise(measurements.reshape(1, -1), 
                                                noise_level=noise_level).flatten()
        
        temporal_measurements.append(measurements.tolist())
        time_points.append(float(idx))  # Assuming 1 sample = 1 hour
    
    return temporal_measurements, time_points, np.array(sampled_health)


# =======================================================
# --- PLOTTING ---
# =======================================================
def plot_convergence(ga, figsize=(12, 5)):
    """Plot fitness convergence curves."""
    fig, axes = plt.subplots(1, 2, figsize=figsize)
    
    # Best fitness
    axes[0].plot(ga.best_fitness_history, linewidth=2.5, color='royalblue', label='Best')
    axes[0].plot(ga.mean_fitness_history, linewidth=2, color='coral', alpha=0.7, label='Mean')
    axes[0].set_xlabel('Generation', fontsize=11)
    axes[0].set_ylabel('Fitness', fontsize=11)
    axes[0].set_title('Fitness Convergence', fontsize=12, fontweight='bold')
    axes[0].grid(True, alpha=0.3)
    axes[0].legend()
    
    # Fitness improvement rate
    improvement = np.diff(ga.best_fitness_history)
    axes[1].plot(improvement, linewidth=2, color='green')
    axes[1].axhline(y=0, color='red', linestyle='--', alpha=0.5)
    axes[1].set_xlabel('Generation', fontsize=11)
    axes[1].set_ylabel('Fitness Improvement', fontsize=11)
    axes[1].set_title('Improvement Rate', fontsize=12, fontweight='bold')
    axes[1].grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()


def plot_degradation_trajectory(ga, best_solution, true_trajectory=None, figsize=(12, 6)):
    """
    Plot predicted vs actual degradation over time.
    """
    n_params = ga.n_health_params
    param_names = ['η_comp', 'η_turb', 'η_comb'][:n_params]
    
    fig, axes = plt.subplots(1, n_params, figsize=figsize)
    if n_params == 1:
        axes = [axes]
    
    for i in range(n_params):
        # Predicted trajectory
        predicted = []
        for time in ga.time_points:
            health = ga.predict_health_at_time(best_solution, time)
            predicted.append(health[i])
        
        axes[i].plot(ga.time_points, predicted, 'o-', linewidth=2.5, 
                    markersize=8, label='Predicted', color='royalblue')
        
        # True trajectory if available
        if true_trajectory is not None:
            true_vals = [true_trajectory(t)[i] for t in ga.time_points]
            axes[i].plot(ga.time_points, true_vals, 's--', linewidth=2, 
                        markersize=7, label='True', color='red', alpha=0.7)
        
        axes[i].set_xlabel('Operating Hours', fontsize=11)
        axes[i].set_ylabel(param_names[i], fontsize=11)
        axes[i].set_title(f'{param_names[i]} Degradation', fontsize=12, fontweight='bold')
        axes[i].grid(True, alpha=0.3)
        axes[i].legend()
    
    plt.tight_layout()
    plt.show()


def plot_measurement_fit(ga, best_solution, figsize=(14, 4)):
    """
    Plot simulated vs actual measurements at each time point.
    """
    n_times = len(ga.time_points)
    fig, axes = plt.subplots(1, n_times, figsize=figsize, sharey=True)
    if n_times == 1:
        axes = [axes]
    
    for t_idx, time in enumerate(ga.time_points):
        # Get simulated and actual measurements
        health = ga.predict_health_at_time(best_solution, time)
        sim_result = list(ga.simulator([health], S=ga.sensors))
        simulated = np.array(sim_result[0]).flatten()
        target = ga.temporal_measurements[t_idx]
        
        x = np.arange(len(ga.sensors))
        width = 0.35
        
        axes[t_idx].bar(x - width/2, target, width, label='Measured', color='coral', alpha=0.8)
        axes[t_idx].bar(x + width/2, simulated, width, label='Simulated', color='royalblue', alpha=0.8)
        axes[t_idx].set_xlabel('Sensors', fontsize=10)
        if t_idx == 0:
            axes[t_idx].set_ylabel('Value', fontsize=11)
        axes[t_idx].set_title(f't = {time:.0f}h', fontsize=11, fontweight='bold')
        axes[t_idx].set_xticks(x)
        axes[t_idx].set_xticklabels(ga.sensors, rotation=45, ha='right')
        axes[t_idx].legend(fontsize=9)
        axes[t_idx].grid(True, alpha=0.3, axis='y')
    
    plt.tight_layout()
    plt.show()
````

````python
import torch
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from typing import Callable, Tuple, List, Optional
from scipy.stats import qmc


class TemporalGeneticAlgorithm:
    """
    Enhanced Genetic Algorithm for temporal degradation tracking.
    
    Features:
    - Multiple measurements over time
    - Degradation modeling (linear/exponential)
    - Early stopping
    - Advanced initialization strategies
    - Adaptive crossover
    - Noise-robust fitness evaluation
    """

    def __init__(self,
                 simulator: Callable,
                 temporal_measurements: List[List[float]],
                 time_points: List[float],
                 sensors: List[str],
                 n_health_params: int = 2,
                 degradation_model: str = "linear",
                 pop_size: int = 100,
                 n_generations: int = 100,
                 health_bounds: Optional[List[Tuple[float, float]]] = None,
                 rate_bounds: Optional[List[Tuple[float, float]]] = None,
                 mutation_rate: float = 0.1,
                 crossover_rate: float = 0.9,
                 tournament_size: int = 3,
                 n_elite: int = 10,
                 mutation_std_ratio: float = 0.05,
                 init_strategy: str = "hybrid",
                 time_weighting: str = "recent",
                 noise_robust: bool = True):
        """
        Parameters
        ----------
        temporal_measurements : list of lists
            Measurements at each time point, shape [n_timesteps, n_sensors]
        time_points : list of float
            Operating hours for each measurement [0, 1000, 2000, ...]
        n_health_params : int
            Number of health parameters (e.g., 2 for [η_comp, η_turb])
        degradation_model : str
            "linear" or "exponential" degradation
        health_bounds : list of tuple
            Bounds for initial health parameters
        rate_bounds : list of tuple
            Bounds for degradation rates
        init_strategy : str
            "random", "latin_hypercube", "informed", or "hybrid"
        time_weighting : str
            "equal", "recent", or "linear" - how to weight different time points
        noise_robust : bool
            Use median-based error metric for outlier resistance
        """
        self.simulator = simulator
        self.temporal_measurements = np.array(temporal_measurements)
        self.time_points = np.array(time_points)
        self.sensors = sensors
        self.n_sensors = len(sensors)
        self.n_health_params = n_health_params
        self.degradation_model = degradation_model
        self.n_timesteps = len(time_points)
        
        # Total genes = initial health params + degradation rates
        self.n_genes = n_health_params * 2
        
        self.pop_size = pop_size
        self.n_generations = n_generations
        self.mutation_rate = mutation_rate
        self.crossover_rate = crossover_rate
        self.tournament_size = tournament_size
        self.n_elite = n_elite
        self.mutation_std_ratio = mutation_std_ratio
        self.init_strategy = init_strategy
        self.time_weighting = time_weighting
        self.noise_robust = noise_robust

        # Set bounds
        if health_bounds is None:
            self.health_bounds = [(0.85, 1.02)] * n_health_params
        else:
            self.health_bounds = health_bounds
            
        if rate_bounds is None:
            # Default: 0 to 0.00005 per hour (5% degradation over 1000 hours)
            self.rate_bounds = [(0.0, 0.00005)] * n_health_params
        else:
            self.rate_bounds = rate_bounds
        
        # Combined bounds: [health_params, degradation_rates]
        self.bounds = self.health_bounds + self.rate_bounds

        # Compute time weights once
        self.time_weights = self._compute_time_weights()

        # Storage
        self.best_fitness_history = []
        self.mean_fitness_history = []
        self.population_history = []
        self.best_solution_history = []

    def _compute_time_weights(self) -> np.ndarray:
        """Compute weights for different time points."""
        weights = np.ones(self.n_timesteps)
        
        if self.time_weighting == "recent":
            # Exponential decay: recent data more important
            max_time = self.time_points[-1]
            weights = np.exp(-0.0005 * self.time_points / max_time)
        elif self.time_weighting == "linear":
            # Linear increase: later data more important
            weights = np.linspace(0.5, 1.5, self.n_timesteps)
        # else: equal weighting (all 1.0)
        
        return weights / weights.sum()  # Normalize

    # =======================================================
    # --- HEALTH PREDICTION ---
    # =======================================================
    def predict_health_at_time(self, individual: torch.Tensor, time: float) -> List[float]:
        """
        Predict health parameters at given time using degradation model.
        
        individual structure: [η_comp_0, η_turb_0, ..., rate_comp, rate_turb, ...]
        """
        initial_health = individual[:self.n_health_params]
        degradation_rates = individual[self.n_health_params:]
        
        if self.degradation_model == "linear":
            # η(t) = η_0 - rate * t
            current_health = initial_health - degradation_rates * time
        elif self.degradation_model == "exponential":
            # η(t) = η_0 * exp(-rate * t)
            current_health = initial_health * torch.exp(-degradation_rates * time)
        else:
            raise ValueError(f"Unknown degradation model: {self.degradation_model}")
        
        # Clamp to physical bounds
        for i in range(self.n_health_params):
            min_val, max_val = self.health_bounds[i]
            current_health[i] = torch.clamp(current_health[i], min_val, max_val)
        
        return current_health.tolist()

    # =======================================================
    # --- FITNESS FUNCTION ---
    # =======================================================
    def fitness(self, individual: torch.Tensor) -> float:
        """
        Evaluate fitness across ALL time points.
        Lower total error = higher fitness.
        """
        total_error = 0.0
        
        for t_idx, time in enumerate(self.time_points):
            # Predict health at this time
            current_health = self.predict_health_at_time(individual, time)
            
            # Simulate measurements with current health
            try:
                sim_result = list(self.simulator([current_health], S=self.sensors))
                simulated = np.array(sim_result[0]).flatten()
            except Exception as e:
                # Penalize invalid simulations
                return -1e6
            
            # Get target measurements
            target = self.temporal_measurements[t_idx]
            
            # Compute errors
            if self.noise_robust:
                # Median absolute percentage error (robust to outliers)
                rel_errors = np.abs(simulated - target) / (np.abs(target) + 1e-8)
                time_error = np.median(rel_errors)
            else:
                # Mean absolute percentage error
                rel_errors = np.abs(simulated - target) / (np.abs(target) + 1e-8)
                time_error = np.mean(rel_errors)
            
            # Weight by time importance
            total_error += time_error * self.time_weights[t_idx]
        
        return -total_error * 1000  # Maximize fitness

    def evaluate_population(self, population: torch.Tensor) -> torch.Tensor:
        """Compute fitness for all individuals."""
        return torch.tensor([self.fitness(ind) for ind in population])

    # =======================================================
    # --- INITIALIZATION ---
    # =======================================================
    def initialize_population(self) -> torch.Tensor:
        """
        Initialize population using selected strategy.
        """
        population = torch.zeros((self.pop_size, self.n_genes))
        
        if self.init_strategy == "random":
            population = self._init_random()
        elif self.init_strategy == "latin_hypercube":
            population = self._init_latin_hypercube()
        elif self.init_strategy == "informed":
            population = self._init_informed()
        elif self.init_strategy == "hybrid":
            # 40% Latin hypercube, 40% informed, 20% random
            n_lhs = int(0.4 * self.pop_size)
            n_informed = int(0.4 * self.pop_size)
            n_random = self.pop_size - n_lhs - n_informed
            
            population[:n_lhs] = self._init_latin_hypercube()[:n_lhs]
            population[n_lhs:n_lhs+n_informed] = self._init_informed()[:n_informed]
            population[n_lhs+n_informed:] = self._init_random()[:n_random]
        else:
            raise ValueError(f"Unknown init strategy: {self.init_strategy}")
        
        return population

    def _init_random(self) -> torch.Tensor:
        """Uniform random initialization."""
        population = torch.zeros((self.pop_size, self.n_genes))
        for i, (min_val, max_val) in enumerate(self.bounds):
            population[:, i] = torch.rand(self.pop_size) * (max_val - min_val) + min_val
        return population

    def _init_latin_hypercube(self) -> torch.Tensor:
        """Latin Hypercube Sampling for better space coverage."""
        sampler = qmc.LatinHypercube(d=self.n_genes)
        samples = sampler.random(n=self.pop_size)
        
        population = torch.zeros((self.pop_size, self.n_genes))
        for i, (min_val, max_val) in enumerate(self.bounds):
            population[:, i] = torch.tensor(samples[:, i] * (max_val - min_val) + min_val)
        
        return population

    def _init_informed(self) -> torch.Tensor:
        """Domain-informed initialization."""
        population = torch.zeros((self.pop_size, self.n_genes))
        
        for i in range(self.pop_size):
            # Initial health: bias towards healthy (95-100%)
            for j in range(self.n_health_params):
                min_val, max_val = self.health_bounds[j]
                population[i, j] = torch.rand(1) * 0.05 * (max_val - min_val) + 0.95 * max_val
            
            # Degradation rates: small positive values
            for j in range(self.n_health_params):
                min_val, max_val = self.rate_bounds[j]
                population[i, self.n_health_params + j] = torch.rand(1) * (max_val - min_val) * 0.5 + min_val
        
        return population

    # =======================================================
    # --- SELECTION ---
    # =======================================================
    def selection(self, population: torch.Tensor, fitness_values: torch.Tensor, n_parents: int) -> torch.Tensor:
        """Tournament selection."""
        parents = torch.zeros((n_parents, self.n_genes))
        for i in range(n_parents):
            candidates_idx = torch.randint(0, self.pop_size, (self.tournament_size,))
            best_idx = candidates_idx[torch.argmax(fitness_values[candidates_idx])]
            parents[i] = population[best_idx]
        return parents

    # =======================================================
    # --- CROSSOVER ---
    # =======================================================
    def adaptive_crossover(self, parent1: torch.Tensor, parent2: torch.Tensor, generation: int) -> Tuple[torch.Tensor, torch.Tensor]:
        """
        Adaptive crossover that adjusts based on generation progress.
        """
        if torch.rand(1).item() > self.crossover_rate:
            return parent1.clone(), parent2.clone()

        progress = generation / self.n_generations
        
        if progress < 0.5:
            # Early generations: exploration (uniform)
            alpha = torch.rand(1).item()
        else:
            # Later generations: exploitation (favor better parent)
            fitness1 = self.fitness(parent1)
            fitness2 = self.fitness(parent2)
            alpha = 0.7 if fitness1 > fitness2 else 0.3

        child1 = alpha * parent1 + (1 - alpha) * parent2
        child2 = (1 - alpha) * parent1 + alpha * parent2

        # Clamp to bounds
        for i, (min_val, max_val) in enumerate(self.bounds):
            child1[i] = torch.clamp(child1[i], min_val, max_val)
            child2[i] = torch.clamp(child2[i], min_val, max_val)

        return child1, child2

    # =======================================================
    # --- MUTATION ---
    # =======================================================
    def mutate(self, individual: torch.Tensor) -> torch.Tensor:
        """Gaussian mutation with adaptive step size."""
        mutated = individual.clone()
        for i, (min_val, max_val) in enumerate(self.bounds):
            if torch.rand(1).item() < self.mutation_rate:
                noise_std = (max_val - min_val) * self.mutation_std_ratio
                mutated[i] += torch.normal(0, noise_std, (1,)).item()
                mutated[i] = torch.clamp(mutated[i], min_val, max_val)
        return mutated

    # =======================================================
    # --- EVOLUTION LOOP ---
    # =======================================================
    def evolve(self, 
               verbose_interval: int = 10,
               patience: int = 30,
               min_improvement: float = 1e-4) -> Tuple[torch.Tensor, dict]:
        """
        Main evolution loop with early stopping.
        
        Returns
        -------
        best_solution : torch.Tensor
        history : dict with convergence metrics
        """
        population = self.initialize_population()
        
        best_fitness_ever = -np.inf
        generations_without_improvement = 0
        
        print(f"\n{'='*70}")
        print(f"Starting Temporal GA Evolution")
        print(f"{'='*70}")
        print(f"Time points: {self.time_points.tolist()}")
        print(f"Degradation model: {self.degradation_model}")
        print(f"Initialization: {self.init_strategy}")
        print(f"{'='*70}\n")

        for generation in range(self.n_generations):
            fitness_values = self.evaluate_population(population)
            
            best_fitness = torch.max(fitness_values).item()
            mean_fitness = torch.mean(fitness_values).item()
            best_idx = torch.argmax(fitness_values)
            best_individual = population[best_idx]
            
            # Record history
            self.best_fitness_history.append(best_fitness)
            self.mean_fitness_history.append(mean_fitness)
            self.population_history.append(population.clone())
            self.best_solution_history.append(best_individual.clone())

            # Early stopping check
            if best_fitness - best_fitness_ever > min_improvement:
                best_fitness_ever = best_fitness
                generations_without_improvement = 0
            else:
                generations_without_improvement += 1

            if generations_without_improvement >= patience:
                print(f"\n{'='*70}")
                print(f"Early stopping at generation {generation}")
                print(f"No improvement for {patience} generations")
                print(f"{'='*70}\n")
                break

            # Elitism
            elite_idx = torch.argsort(fitness_values, descending=True)[:self.n_elite]
            elite = population[elite_idx]

            # Selection
            n_parents = self.pop_size - self.n_elite
            parents = self.selection(population, fitness_values, n_parents)

            # Create new generation
            new_population = torch.zeros((self.pop_size, self.n_genes))
            new_population[:self.n_elite] = elite

            # Crossover + Mutation
            for i in range(self.n_elite, self.pop_size, 2):
                parent1 = parents[torch.randint(0, n_parents, (1,)).item()]
                parent2 = parents[torch.randint(0, n_parents, (1,)).item()]
                child1, child2 = self.adaptive_crossover(parent1, parent2, generation)
                new_population[i] = self.mutate(child1)
                if i + 1 < self.pop_size:
                    new_population[i + 1] = self.mutate(child2)

            population = new_population

            # Logging
            if generation % verbose_interval == 0 or generation == self.n_generations - 1:
                print(f"Gen {generation:4d} | Best: {best_fitness:8.2f} | Mean: {mean_fitness:8.2f} | Genes: {best_individual.tolist()}")

        # Final evaluation
        print(f"\n{'='*70}")
        print("Evolution Complete - Final Results")
        print(f"{'='*70}")
        
        final_fitness = self.evaluate_population(population)
        best_idx = torch.argmax(final_fitness)
        best_solution = population[best_idx]
        
        # Display predicted degradation trajectory
        print(f"\nBest Solution:")
        print(f"  Initial Health: {best_solution[:self.n_health_params].tolist()}")
        print(f"  Degradation Rates: {best_solution[self.n_health_params:].tolist()}")
        
        print(f"\nPredicted Health Trajectory:")
        for t_idx, time in enumerate(self.time_points):
            health = self.predict_health_at_time(best_solution, time)
            print(f"  t={time:6.0f}h: {health}")
            
            # Show measurement comparison
            sim_result = list(self.simulator([health], S=self.sensors))
            simulated = np.array(sim_result[0]).flatten()
            target = self.temporal_measurements[t_idx]
            errors = np.abs(simulated - target) / np.abs(target) * 100
            print(f"           Simulated: {simulated}")
            print(f"           Target:    {target}")
            print(f"           Error (%): {errors}\n")
        
        history = {
            'best_fitness': self.best_fitness_history,
            'mean_fitness': self.mean_fitness_history,
            'population': self.population_history,
            'best_solutions': self.best_solution_history
        }
        
        return best_solution, history


# =======================================================
# --- UTILITY FUNCTIONS ---
# =======================================================
def add_measurement_noise(measurements: np.ndarray, 
                         noise_level: float = 0.02,
                         outlier_prob: float = 0.05) -> np.ndarray:
    """
    Add realistic sensor noise to measurements.
    
    Parameters
    ----------
    measurements : array
        Clean measurements
    noise_level : float
        Gaussian noise standard deviation (relative)
    outlier_prob : float
        Probability of outlier for each measurement
    """
    noisy = measurements.copy()
    
    # Gaussian noise
    noise = np.random.normal(0, noise_level, measurements.shape)
    noisy = noisy * (1 + noise)
    
    # Occasional outliers
    outlier_mask = np.random.rand(*measurements.shape) < outlier_prob
    if outlier_mask.any():
        outliers = np.random.uniform(-0.1, 0.1, outlier_mask.sum())
        noisy[outlier_mask] *= (1 + outliers)
    
    return noisy


def prepare_temporal_data(true_trajectory, simulator, sensors, time_indices, noise_level=0.0):
    """
    Helper function to prepare temporal measurements from a degradation trajectory.
    
    Parameters
    ----------
    true_trajectory : array-like
        Full degradation trajectory, shape [n_samples, n_health_params]
    simulator : callable
        Simulator function
    sensors : list of str
        Sensor names
    time_indices : list of int
        Indices to sample from trajectory (e.g., [0, 10, 20, 30])
    noise_level : float
        Noise level to add (0.0 = no noise)
    
    Returns
    -------
    temporal_measurements : list of lists
        Measurements at each time point
    time_points : list of float
        Corresponding time points (in hours, assuming 1 sample = 1 hour)
    sampled_health : array
        True health states at sampled time points
    """
    temporal_measurements = []
    time_points = []
    sampled_health = []
    
    for idx in time_indices:
        # Get true health at this time
        health = true_trajectory[idx]
        sampled_health.append(health)
        
        # Simulate measurements
        sim_result = list(simulator([health], S=sensors))
        measurements = np.array(sim_result[0]).flatten()
        
        # Add noise if requested
        if noise_level > 0:
            measurements = add_measurement_noise(measurements.reshape(1, -1), 
                                                noise_level=noise_level).flatten()
        
        temporal_measurements.append(measurements.tolist())
        time_points.append(float(idx))  # Assuming 1 sample = 1 hour
    
    return temporal_measurements, time_points, np.array(sampled_health)


# =======================================================
# --- PLOTTING ---
# =======================================================
def plot_convergence(ga, figsize=(12, 5)):
    """Plot fitness convergence curves."""
    fig, axes = plt.subplots(1, 2, figsize=figsize)
    
    # Best fitness
    axes[0].plot(ga.best_fitness_history, linewidth=2.5, color='royalblue', label='Best')
    axes[0].plot(ga.mean_fitness_history, linewidth=2, color='coral', alpha=0.7, label='Mean')
    axes[0].set_xlabel('Generation', fontsize=11)
    axes[0].set_ylabel('Fitness', fontsize=11)
    axes[0].set_title('Fitness Convergence', fontsize=12, fontweight='bold')
    axes[0].grid(True, alpha=0.3)
    axes[0].legend()
    
    # Fitness improvement rate
    improvement = np.diff(ga.best_fitness_history)
    axes[1].plot(improvement, linewidth=2, color='green')
    axes[1].axhline(y=0, color='red', linestyle='--', alpha=0.5)
    axes[1].set_xlabel('Generation', fontsize=11)
    axes[1].set_ylabel('Fitness Improvement', fontsize=11)
    axes[1].set_title('Improvement Rate', fontsize=12, fontweight='bold')
    axes[1].grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()


def plot_degradation_trajectory(ga, best_solution, true_trajectory=None, figsize=(12, 6)):
    """
    Plot predicted vs actual degradation over time.
    """
    n_params = ga.n_health_params
    param_names = ['η_comp', 'η_turb', 'η_comb'][:n_params]
    
    fig, axes = plt.subplots(1, n_params, figsize=figsize)
    if n_params == 1:
        axes = [axes]
    
    for i in range(n_params):
        # Predicted trajectory
        predicted = []
        for time in ga.time_points:
            health = ga.predict_health_at_time(best_solution, time)
            predicted.append(health[i])
        
        axes[i].plot(ga.time_points, predicted, 'o-', linewidth=2.5, 
                    markersize=8, label='Predicted', color='royalblue')
        
        # True trajectory if available
        if true_trajectory is not None:
            true_vals = [true_trajectory(t)[i] for t in ga.time_points]
            axes[i].plot(ga.time_points, true_vals, 's--', linewidth=2, 
                        markersize=7, label='True', color='red', alpha=0.7)
        
        axes[i].set_xlabel('Operating Hours', fontsize=11)
        axes[i].set_ylabel(param_names[i], fontsize=11)
        axes[i].set_title(f'{param_names[i]} Degradation', fontsize=12, fontweight='bold')
        axes[i].grid(True, alpha=0.3)
        axes[i].legend()
    
    plt.tight_layout()
    plt.show()


def plot_measurement_fit(ga, best_solution, figsize=(14, 4)):
    """
    Plot simulated vs actual measurements at each time point.
    """
    n_times = len(ga.time_points)
    fig, axes = plt.subplots(1, n_times, figsize=figsize, sharey=True)
    if n_times == 1:
        axes = [axes]
    
    for t_idx, time in enumerate(ga.time_points):
        # Get simulated and actual measurements
        health = ga.predict_health_at_time(best_solution, time)
        sim_result = list(ga.simulator([health], S=ga.sensors))
        simulated = np.array(sim_result[0]).flatten()
        target = ga.temporal_measurements[t_idx]
        
        x = np.arange(len(ga.sensors))
        width = 0.35
        
        axes[t_idx].bar(x - width/2, target, width, label='Measured', color='coral', alpha=0.8)
        axes[t_idx].bar(x + width/2, simulated, width, label='Simulated', color='royalblue', alpha=0.8)
        axes[t_idx].set_xlabel('Sensors', fontsize=10)
        if t_idx == 0:
            axes[t_idx].set_ylabel('Value', fontsize=11)
        axes[t_idx].set_title(f't = {time:.0f}h', fontsize=11, fontweight='bold')
        axes[t_idx].set_xticks(x)
        axes[t_idx].set_xticklabels(ga.sensors, rotation=45, ha='right')
        axes[t_idx].legend(fontsize=9)
        axes[t_idx].grid(True, alpha=0.3, axis='y')
    
    plt.tight_layout()
    plt.show()
````

````python
# ==========================================================
# Example: Single Measurement (3 genes only, no degradation)
# ==========================================================
import time
import numpy as np

# ----------------------------------------------------------
# STEP 1 - Simulate True Health & Target Measurements
# ----------------------------------------------------------
true_traj = simulate_degradation_trajectory(speed='fast', waterwash='rare', size=100, seed=0)
sensors = ['P3', 'T5', 'Wf']

# Get simulated sensor measurements for one time step
measured_series = list(h_SCSF(true_traj, S=sensors))
target_measurements = measured_series[10]  # 10th sample

print(f"True health: {true_traj[10]}")
print(f"Target measurements: {target_measurements}")

# ----------------------------------------------------------
# STEP 2 — Create and Run GA (3 genes only)
# ----------------------------------------------------------
start_time = time.time()

ga = TemporalGeneticAlgorithm(
    simulator=h_SCSF,
    temporal_measurements=[target_measurements],  # Single measurement
    time_points=[10.0],                           # Single time point
    sensors=sensors,
    n_health_params=3,                            # η_comp, η_turb, η_comb
    estimate_degradation=False,                   # ← KEY: Only estimate health, not rates
    pop_size=10,
    n_generations=10,
    health_bounds=[(0.95, 1.02), (0.95, 1.02)],
    mutation_rate=0.1,
    crossover_rate=0.9,
    tournament_size=3,
    n_elite=15,
    mutation_std_ratio=0.05,
    init_strategy="hybrid"  # Better than random
)

best_solution, history = ga.evolve(verbose_interval=5, patience=20)

print("\n--- End of Evolution ---")
print(f"Execution time: {time.time() - start_time:.2f} s")

# ----------------------------------------------------------
# STEP 3 - Display Best Solution
# ----------------------------------------------------------
eta_comp_est, eta_turb_est, eta_comb_est = best_solution.cpu().numpy()
true_eta_comp, true_eta_turb = true_traj[10]

print(f"\nEstimated health: η_comp={eta_comp_est:.4f}, η_turb={eta_turb_est:.4f}, η_comb={eta_comb_est:.4f}")
print(f"True health:      η_comp={true_eta_comp:.4f}, η_turb={true_eta_turb:.4f}, η_comb={true_eta_comb:.4f}")

# Compute absolute errors
error_comp = abs(true_eta_comp - eta_comp_est)
error_turb = abs(true_eta_turb - eta_turb_est)
# error_comb = abs(true_eta_comb - eta_comb_est)

print(f"Error on η_comp: {error_comp:.6f}")
print(f"Error on η_turb: {error_turb:.6f}")
print(f"Error on η_comb: {error_comb:.6f}")

# ----------------------------------------------------------
# STEP 4 — Visualizations
# ----------------------------------------------------------
plot_convergence(ga)
# Note: plot_ga_population needs modification for the new class
````
