```python
import torch
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import pickle
import time
from typing import List, Tuple, Callable, Dict, Any
from scipy.stats import qmc  # For Latin Hypercube Sampling

# ==============================================================================
# MODULE 1: SIMULATOR WRAPPER (Robustness)
# ==============================================================================
class SimulatorInterface:
    def __init__(self, simulator_func: Callable, sensors: List[str]):
        self.sim_func = simulator_func
        self.sensors = sensors

    def safe_evaluate(self, individual: np.ndarray) -> np.ndarray:
        """
        Wraps the simulator to handle crashes (Physical Constraints).
        Returns None if the simulation fails (NaN, Sqrt error, etc).
        """
        try:
            # Assumes simulator takes list of lists/arrays
            res = list(self.sim_func([individual], S=self.sensors))
            measurements = np.array(res[0]).flatten()
            
            # Check for NaNs
            if np.isnan(measurements).any():
                return None
            return measurements
        except Exception as e:
            # Log error if needed, but do not crash
            return None

# ==============================================================================
# MODULE 2: METRICS LOGGER (Observability)
# ==============================================================================
class GAMetrics:
    def __init__(self):
        self.history = {
            "generation": [],
            "best_fitness": [],
            "avg_fitness": [],
            "pop_diversity_std": [], # Average Std Dev across genes
            "best_genes": [],
            "timestamp": []
        }
        self.start_time = time.time()

    def log(self, gen, fitnesses, population):
        # Calculate Diversity (Mean Standard Deviation across dimensions)
        std_per_gene = torch.std(population, dim=0)
        avg_std = torch.mean(std_per_gene).item()
        
        best_idx = torch.argmax(fitnesses)
        
        self.history["generation"].append(gen)
        self.history["best_fitness"].append(fitnesses[best_idx].item())
        self.history["avg_fitness"].append(torch.mean(fitnesses).item())
        self.history["pop_diversity_std"].append(avg_std)
        self.history["best_genes"].append(population[best_idx].cpu().numpy())
        self.history["timestamp"].append(time.time() - self.start_time)

    def save(self, filename="ga_results.pkl"):
        with open(filename, "wb") as f:
            pickle.dump(self.history, f)
        print(f"Results saved to {filename}")

# ==============================================================================
# MODULE 3: THE OPTIMIZED GENETIC ALGORITHM
# ==============================================================================
class RobustGA:
    def __init__(self, 
                 simulator: SimulatorInterface,
                 target: np.ndarray,
                 bounds: List[Tuple[float, float]],
                 pop_size: int = 100,
                 n_generations: int = 50,
                 mutation_rate_start: float = 0.2,
                 patience: int = 10):
        
        self.sim = simulator
        self.target = torch.tensor(target, dtype=torch.float32)
        self.bounds = bounds
        self.n_genes = len(bounds)
        self.pop_size = pop_size
        self.n_generations = n_generations
        
        # Adaptive Mutation Parameters
        self.mut_rate_start = mutation_rate_start
        self.current_mut_rate = mutation_rate_start
        
        # Early Stopping
        self.patience = patience
        self.best_ever_fitness = -float('inf')
        self.generations_without_improve = 0
        
        self.logger = GAMetrics()

    def initialize_population_lhs(self) -> torch.Tensor:
        """
        Uses Latin Hypercube Sampling for better initial coverage.
        """
        sampler = qmc.LatinHypercube(d=self.n_genes)
        sample = sampler.random(n=self.pop_size) # [0, 1]
        
        # Scale to bounds
        population = torch.tensor(sample, dtype=torch.float32)
        for i, (low, high) in enumerate(self.bounds):
            population[:, i] = population[:, i] * (high - low) + low
            
        return population

    def calculate_fitness_batch(self, population: torch.Tensor) -> torch.Tensor:
        """
        Calculates fitness. Uses a Penalty for invalid simulations.
        """
        fitnesses = []
        penalty_value = -1e6 # Large negative number
        
        # TODO: If simulator supports true batching, vectorize this.
        # Otherwise, we loop.
        for ind in population:
            genes_np = ind.cpu().numpy()
            sim_measurements = self.sim.safe_evaluate(genes_np)
            
            if sim_measurements is None:
                fitnesses.append(penalty_value)
            else:
                sim_tensor = torch.tensor(sim_measurements, dtype=torch.float32)
                # Relative Error (MAPE) approach
                # Add epsilon to target to avoid division by zero
                rel_error = torch.abs(sim_tensor - self.target) / (torch.abs(self.target) + 1e-8)
                mean_error = torch.mean(rel_error)
                
                # Fitness = negative error (maximization problem)
                fitnesses.append(-mean_error.item() * 1000)
                
        return torch.tensor(fitnesses, dtype=torch.float32)

    def adaptive_mutation(self, individual: torch.Tensor, progress: float) -> torch.Tensor:
        """
        Decays mutation strength as generations progress.
        progress: 0.0 (start) to 1.0 (end)
        """
        mutated = individual.clone()
        
        # Decay mutation probability and strength
        current_rate = self.mut_rate_start * (1 - progress * 0.8) # Decays to 20% of original
        
        for i, (low, high) in enumerate(self.bounds):
            if torch.rand(1).item() < current_rate:
                # Dynamic Sigma: Search range * 10% * decay factor
                sigma = (high - low) * 0.1 * (1 - progress) 
                noise = torch.normal(0, sigma, (1,)).item()
                mutated[i] += noise
                mutated[i] = torch.clamp(mutated[i], low, high)
        return mutated

    def crossover(self, p1: torch.Tensor, p2: torch.Tensor) -> Tuple[torch.Tensor, torch.Tensor]:
        """Uniform Crossover"""
        alpha = torch.rand(1).item()
        c1 = alpha * p1 + (1 - alpha) * p2
        c2 = (1 - alpha) * p1 + alpha * p2
        return c1, c2

    def run(self, seed_population=None):
        print(f"--- Starting GA Optimization ({self.n_genes} components) ---")
        
        # 1. Initialization
        if seed_population is not None:
            # Use provided seed (e.g., from previous flight)
            population = seed_population
            print("-> Initialized with seeded population.")
        else:
            population = self.initialize_population_lhs()
            print("-> Initialized with Latin Hypercube Sampling.")

        for gen in range(self.n_generations):
            # 2. Evaluation
            fitnesses = self.calculate_fitness_batch(population)
            
            # 3. Logging
            self.logger.log(gen, fitnesses, population)
            
            current_best = torch.max(fitnesses).item()
            
            # 4. Early Stopping Logic
            if current_best > self.best_ever_fitness + 1e-4: # Tolerance
                self.best_ever_fitness = current_best
                self.generations_without_improve = 0
            else:
                self.generations_without_improve += 1
                
            if self.generations_without_improve >= self.patience:
                print(f"Stopping early at Gen {gen}: No improvement for {self.patience} generations.")
                break

            # 5. Selection (Tournament) & Elitism
            n_elite = max(2, int(self.pop_size * 0.05)) # Top 5%
            sorted_indices = torch.argsort(fitnesses, descending=True)
            elite_pop = population[sorted_indices[:n_elite]]
            
            # Select parents
            new_pop = [elite_pop]
            
            while sum([len(b) for b in new_pop]) < self.pop_size:
                # Tournament
                candidates = torch.randint(0, self.pop_size, (3,))
                p1_idx = candidates[torch.argmax(fitnesses[candidates])]
                candidates = torch.randint(0, self.pop_size, (3,))
                p2_idx = candidates[torch.argmax(fitnesses[candidates])]
                
                # Crossover
                c1, c2 = self.crossover(population[p1_idx], population[p2_idx])
                
                # Adaptive Mutation
                progress = gen / self.n_generations
                c1 = self.adaptive_mutation(c1, progress)
                c2 = self.adaptive_mutation(c2, progress)
                
                # Append (handling tensor shapes)
                new_pop.append(c1.unsqueeze(0))
                if sum([len(b) for b in new_pop]) < self.pop_size:
                    new_pop.append(c2.unsqueeze(0))
            
            population = torch.cat(new_pop, dim=0)
            
            # Clamp boundaries finally
            for i, (low, high) in enumerate(self.bounds):
                population[:, i] = torch.clamp(population[:, i], low, high)

            if gen % 10 == 0:
                print(f"Gen {gen} | Fitness: {current_best:.4f} | Diversity: {self.logger.history['pop_diversity_std'][-1]:.4f}")

        # Return best solution and full logs
        final_fitnesses = self.calculate_fitness_batch(population)
        best_idx = torch.argmax(final_fitnesses)
        return population[best_idx], self.logger

# ==============================================================================
# MODULE 4: VISUALIZATION
# ==============================================================================
def plot_dashboard(metrics: GAMetrics, bounds, true_val=None):
    hist = metrics.history
    
    fig = plt.figure(figsize=(15, 10))
    
    # 1. Convergence
    ax1 = fig.add_subplot(2, 2, 1)
    ax1.plot(hist['generation'], hist['best_fitness'], label='Best', color='blue')
    ax1.plot(hist['generation'], hist['avg_fitness'], label='Average', color='orange', alpha=0.6)
    ax1.set_title("Fitness Convergence")
    ax1.set_xlabel("Generation")
    ax1.set_ylabel("Fitness (-MAPE)")
    ax1.legend()
    ax1.grid(True, alpha=0.3)
    
    # 2. Diversity
    ax2 = fig.add_subplot(2, 2, 2)
    ax2.plot(hist['generation'], hist['pop_diversity_std'], color='green')
    ax2.set_title("Population Diversity (Std Dev)")
    ax2.set_xlabel("Generation")
    ax2.set_ylabel("Avg Std Dev")
    ax2.grid(True, alpha=0.3)
    
    # 3. Component Trajectory (for 1st component)
    best_genes_np = np.array(hist['best_genes'])
    ax3 = fig.add_subplot(2, 2, 3)
    ax3.plot(hist['generation'], best_genes_np[:, 0], color='purple')
    if true_val is not None:
        ax3.axhline(true_val[0], color='red', linestyle='--', label='True Value')
    ax3.set_title(f"Component 1 Evolution ({bounds[0]})")
    ax3.set_ylim(bounds[0])
    ax3.legend()
    ax3.grid(True)
    
    plt.tight_layout()
    plt.show()
```

```python
import torch
import numpy as np
import pickle
import time
from typing import List, Tuple, Callable, Dict, Optional

# ==============================================================================
# MODULE 1: DEGRADATION MODEL (The Decoder)
# ==============================================================================
class DegradationModel:
    """
    Decodes the abstract 'Genes' into physical Health Trajectories.
    
    Assumes Linear Degradation: 
    Health(t) = Initial_Health - (Degradation_Rate * t)
    
    Gene Structure: 
    [Init_Comp1, Init_Comp2..., Rate_Comp1, Rate_Comp2...]
    Total Genes = 2 * Number of Components
    """
    def __init__(self, n_components: int, n_timesteps: int):
        self.n_comps = n_components
        self.T = n_timesteps

    def genes_to_trajectory(self, genes: torch.Tensor) -> np.ndarray:
        """
        Converts a single individual (Tensor of genes) into a trajectory (T, n_comps)
        """
        # Split genes into Initial Values and Rates
        initials = genes[:self.n_comps] 
        rates = genes[self.n_comps:]    

        # Create time vector [0, 1, 2, ..., T-1]
        t = torch.arange(self.T, device=genes.device)

        # Calculate trajectory: shape (T, n_comps)
        # Broadcasting: (n_comps,) - (n_comps,) * (T, 1) -> Transpose logic needed
        # Result: Row = Time, Col = Component
        trajectory = initials - (rates.unsqueeze(1) * t.unsqueeze(0)).T
        
        return trajectory.cpu().numpy() # Return as numpy for the simulator

# ==============================================================================
# MODULE 2: ROBUST SIMULATOR INTERFACE (The Evaluator)
# ==============================================================================
class RobustSimulator:
    """
    Wraps the physical simulator (h_SCSF) with:
    1. Error Handling (NaN/Crash protection)
    2. Sensor Selection
    3. Noise Injection (for testing robustness)
    """
    def __init__(self, 
                 simulator_func: Callable, 
                 sensors: List[str], 
                 add_sensor_noise: bool = False,
                 noise_std: float = 0.01):
        
        self.sim_func = simulator_func
        self.sensors = sensors
        self.add_noise = add_sensor_noise
        self.noise_std = noise_std

    def evaluate_trajectory(self, health_traj: np.ndarray) -> Optional[np.ndarray]:
        """
        Input: Health Trajectory (T, n_components)
        Output: Sensor Trajectory (T, n_sensors) or None if crash
        """
        try:
            # Call external simulator
            # Assumes h_SCSF takes a batch or loop is handled inside.
            # If h_SCSF expects list of lists:
            res = list(self.sim_func(health_traj, S=self.sensors))
            sensor_data = np.array(res) # Shape (T, n_sensors)

            # Sanity Check for Physics (e.g., negative pressures)
            if np.isnan(sensor_data).any() or np.isinf(sensor_data).any():
                return None
            
            # Optional: Add Noise to simulated data (Simulate dirty sensors)
            if self.add_noise:
                noise = np.random.normal(0, self.noise_std, sensor_data.shape)
                sensor_data += noise

            return sensor_data

        except Exception as e:
            # print(f"Simulator Crash: {e}") # Uncomment for debug
            return None

# ==============================================================================
# MODULE 3: EVOLUTIONARY OPTIMIZER (The Solver)
# ==============================================================================
class EvolutionaryOptimizer:
    def __init__(self, 
                 model: DegradationModel,
                 simulator: RobustSimulator,
                 target_data: np.ndarray,
                 bounds: List[Tuple[float, float]],
                 pop_size: int = 100,
                 n_generations: int = 50,
                 mutation_rate_start: float = 0.3,
                 mutation_rate_end: float = 0.01,
                 crossover_rate: float = 0.9,
                 tournament_size: int = 3):
        
        self.model = model
        self.sim = simulator
        self.target = torch.tensor(target_data, dtype=torch.float32)
        self.bounds = bounds
        self.pop_size = pop_size
        self.n_generations = n_generations
        self.n_genes = len(bounds)
        
        # Hyperparameters
        self.mut_rate_start = mutation_rate_start
        self.mut_rate_end = mutation_rate_end
        self.cross_rate = crossover_rate
        self.tourn_size = tournament_size
        
        # Storage (The "Black Box" Recorder)
        self.history = {
            "fitness_best": [],
            "fitness_avg": [],
            "fitness_std": [], # Diversity metric 1
            "pop_diversity": [], # Diversity metric 2 (Geo distance)
            "best_genes": [],
            "best_trajectory_pred": [] # Store the curve for plotting
        }

    # --- CORE 1: INITIALIZATION (Latin Hypercube approach via Torch) ---
    def initialize_population(self) -> torch.Tensor:
        # Uniform random [0,1]
        pop = torch.rand(self.pop_size, self.n_genes)
        
        # Scale to bounds
        for i, (low, high) in enumerate(self.bounds):
            pop[:, i] = pop[:, i] * (high - low) + low
        return pop

    # --- CORE 2: FITNESS (Normalized RMSE) ---
    def calculate_fitness(self, population: torch.Tensor) -> torch.Tensor:
        fitnesses = []
        penalty = -1e6
        
        for i in range(len(population)):
            genes = population[i]
            
            # 1. Decode Genes -> Health Trajectory
            health_traj = self.model.genes_to_trajectory(genes)
            
            # 2. Simulate -> Sensor Trajectory
            sim_sensors = self.sim.evaluate_trajectory(health_traj)
            
            if sim_sensors is None:
                fitnesses.append(penalty)
                continue
                
            sim_tensor = torch.tensor(sim_sensors, dtype=torch.float32)
            
            # 3. Compute Normalized Error (MAPE-like)
            # We divide by target to normalize scale (Pressure vs Temp)
            # abs(Sim - True) / True
            rel_error = torch.abs(sim_tensor - self.target) / (torch.abs(self.target) + 1e-8)
            
            # Mean error over Time and Sensors
            mean_error = torch.mean(rel_error)
            
            # Fitness is negative error
            fitnesses.append(-mean_error.item())
            
        return torch.tensor(fitnesses)

    # --- CORE 3: ADAPTIVE MUTATION ---
    def mutate(self, individual: torch.Tensor, progress: float) -> torch.Tensor:
        """
        Gaussian Mutation with Annealing.
        progress: 0.0 to 1.0 (Generation / Total_Gen)
        """
        # Linearly interpolate mutation strength
        current_sigma_scale = self.mut_rate_start - (self.mut_rate_start - self.mut_rate_end) * progress
        
        mutated = individual.clone()
        
        # Apply mutation to each gene with probability
        # We mutate ALWAYS but with varying Strength (Evolution Strategy style)
        for i, (low, high) in enumerate(self.bounds):
             # Sigma is % of the bound range
            sigma = (high - low) * current_sigma_scale
            noise = torch.normal(0, sigma, (1,)).item()
            
            mutated[i] += noise
            mutated[i] = torch.clamp(mutated[i], low, high)
            
        return mutated

    # --- CORE 4: BLENDED CROSSOVER ---
    def crossover(self, p1: torch.Tensor, p2: torch.Tensor) -> Tuple[torch.Tensor, torch.Tensor]:
        if torch.rand(1).item() > self.cross_rate:
            return p1.clone(), p2.clone()
        
        # Blended Crossover (alpha=0.5 means average)
        # Child = alpha*P1 + (1-alpha)*P2
        alpha = torch.rand(1).item() 
        
        c1 = alpha * p1 + (1 - alpha) * p2
        c2 = (1 - alpha) * p1 + alpha * p2
        
        return c1, c2

    # --- MAIN LOOP ---
    def run(self):
        print(f"--- GA Started: {self.n_genes} Genes ({self.model.n_comps} Components) ---")
        population = self.initialize_population()
        
        for gen in range(self.n_generations):
            # 1. Evaluate
            fitnesses = self.calculate_fitness(population)
            
            # 2. Metrics & Logging
            best_idx = torch.argmax(fitnesses)
            best_val = fitnesses[best_idx].item()
            std_val = torch.std(fitnesses).item()
            
            # Diversity: Mean Std Dev of Genes across population
            pop_diversity = torch.mean(torch.std(population, dim=0)).item()
            
            self.history["fitness_best"].append(best_val)
            self.history["fitness_avg"].append(torch.mean(fitnesses).item())
            self.history["fitness_std"].append(std_val)
            self.history["pop_diversity"].append(pop_diversity)
            self.history["best_genes"].append(population[best_idx].clone().cpu().numpy())
            
            # Store predicted trajectory for visualization (only for best)
            best_health_traj = self.model.genes_to_trajectory(population[best_idx])
            best_sensor_pred = self.sim.evaluate_trajectory(best_health_traj)
            self.history["best_trajectory_pred"].append(best_sensor_pred)

            if gen % 5 == 0:
                print(f"Gen {gen:3d} | Fitness: {best_val:.5f} | Diversity: {pop_diversity:.4f}")

            # 3. Elitism (Keep top 5%)
            n_elite = max(2, int(self.pop_size * 0.05))
            sorted_indices = torch.argsort(fitnesses, descending=True)
            elite = population[sorted_indices[:n_elite]]
            
            # 4. Create New Population
            new_pop = [elite]
            
            # Generate offspring
            while sum([len(p) for p in new_pop]) < self.pop_size:
                # Tournament Selection
                cands = torch.randint(0, self.pop_size, (self.tourn_size,))
                p1 = population[cands[torch.argmax(fitnesses[cands])]]
                
                cands = torch.randint(0, self.pop_size, (self.tourn_size,))
                p2 = population[cands[torch.argmax(fitnesses[cands])]]
                
                # Crossover
                c1, c2 = self.crossover(p1, p2)
                
                # Mutation (Adaptive)
                progress = gen / self.n_generations
                c1 = self.mutate(c1, progress)
                c2 = self.mutate(c2, progress)
                
                new_pop.append(c1.unsqueeze(0))
                if sum([len(p) for p in new_pop]) < self.pop_size:
                    new_pop.append(c2.unsqueeze(0))
            
            population = torch.cat(new_pop, dim=0)
            
            # Boundary Clamp (Safety)
            for i, (low, high) in enumerate(self.bounds):
                population[:, i] = torch.clamp(population[:, i], low, high)

        return population[torch.argmax(fitnesses)], self.history
```

```python
import torch
import numpy as np
import pickle
import time
from typing import List, Tuple, Callable, Optional
from concurrent.futures import ProcessPoolExecutor
from tqdm import tqdm # Progress bar

# ==============================================================================
# TOP LEVEL WORKER FUNCTION (Must be outside the class for Multiprocessing)
# ==============================================================================
import warnings

def evaluate_worker(args):
    """
    Robust worker that suppresses warnings and catches physics crashes.
    """
    genes, n_comps, T, sim_func, sensors, target = args
    
    # 1. Suppress specific Numpy warnings (hides the "invalid value in sqrt" spam)
    with np.errstate(invalid='ignore', divide='ignore'):
        try:
            # --- DECODE ---
            initials = genes[:n_comps]
            rates = genes[n_comps:]
            t_steps = np.arange(T)
            
            # Trajectory = Init - (Rate * t)
            # Broadcasting fix: (N, 1) - (N, 1) * (1, T) -> (N, T) -> Transpose to (T, N)
            health_traj = initials[:, np.newaxis] - (rates[:, np.newaxis] * t_steps)
            health_traj = health_traj.T 

            # --- SIMULATE ---
            # If the simulator throws a warning, numpy usually produces NaNs 
            # instead of raising an Exception. We must check for NaNs.
            res = list(sim_func(health_traj, S=sensors))
            sim_data = np.array(res)

            # Check for Physics Crash (NaNs or Infinities)
            if np.isnan(sim_data).any() or np.isinf(sim_data).any():
                return -1e9 # "Death Penalty"

            # --- CALCULATE FITNESS ---
            diff = np.abs(sim_data - target)
            # Add epsilon to target to avoid division by zero
            rel_error = diff / (np.abs(target) + 1e-8)
            
            # Fitness = Negative Mean Error
            return -np.mean(rel_error)

        except Exception:
            # Catches Python-level errors (e.g. index out of bounds)
            return -1e9

# ==============================================================================
# OPTIMIZED CLASS
# ==============================================================================
class FastEvolutionaryOptimizer:
    def __init__(self, 
                 simulator_func: Callable,
                 target_data: np.ndarray,
                 sensors: List[str],
                 bounds: List[Tuple[float, float]],
                 pop_size: int = 100,
                 n_generations: int = 50,
                 n_components: int = 2,
                 n_workers: int = 4): # NEW: Number of CPU cores to use
        
        self.sim_func = simulator_func
        self.target = target_data
        self.sensors = sensors
        self.bounds = bounds
        self.pop_size = pop_size
        self.n_generations = n_generations
        self.n_comps = n_components
        self.n_workers = n_workers
        self.n_genes = len(bounds)
        self.T = target_data.shape[0]
        
        # Adaptive Parameters
        self.mut_rate_start = 0.3
        self.mut_rate_end = 0.01

        # History
        self.history = {"fitness_best": [], "best_genes": []}

    def initialize_population(self):
        pop = np.random.rand(self.pop_size, self.n_genes)
        for i, (low, high) in enumerate(self.bounds):
            pop[:, i] = pop[:, i] * (high - low) + low
        return pop

    def run(self):
        print(f"--- Fast GA Started on {self.n_workers} Cores ---")
        
        # Use Numpy for population (Faster than Torch for multiprocessing serialization)
        population = self.initialize_population()
        
        # Main Loop with Progress Bar
        for gen in tqdm(range(self.n_generations), desc="Evolving"):
            
            # --- PARALLEL EVALUATION ---
            # Prepare arguments for every individual
            # We pack everything needed into a tuple for the worker
            tasks = []
            for ind in population:
                tasks.append((
                    ind, 
                    self.n_comps, 
                    self.T, 
                    self.sim_func, 
                    self.sensors, 
                    self.target
                ))

            # Execute in Parallel
            with ProcessPoolExecutor(max_workers=self.n_workers) as executor:
                results = list(executor.map(evaluate_worker, tasks))
            
            fitnesses = np.array(results)
            
            # Logging
            best_idx = np.argmax(fitnesses)
            self.history["fitness_best"].append(fitnesses[best_idx])
            self.history["best_genes"].append(population[best_idx].copy())

            # --- STANDARD GA OPERATIONS (Fast Numpy Vectorized) ---
            # 1. Elitism
            sorted_idx = np.argsort(fitnesses)[::-1]
            n_elite = max(2, int(self.pop_size * 0.05))
            elite = population[sorted_idx[:n_elite]]
            
            # 2. Selection (Vectorized Tournament)
            # Create pool of parents
            parent_indices = np.random.randint(0, self.pop_size, (self.pop_size - n_elite, 3))
            tourn_fitness = fitnesses[parent_indices]
            winners = np.argmax(tourn_fitness, axis=1)
            # Get actual indices of winners
            winner_indices = parent_indices[np.arange(len(winners)), winners]
            parents = population[winner_indices]

            # 3. Crossover & Mutation (Vectorized)
            # Split parents into pairs
            p1 = parents[0::2]
            p2 = parents[1::2]
            
            # Ensure equal length
            min_len = min(len(p1), len(p2))
            p1 = p1[:min_len]
            p2 = p2[:min_len]
            
            # BLX-Alpha Crossover
            alpha = np.random.rand(min_len, 1)
            c1 = alpha * p1 + (1 - alpha) * p2
            c2 = (1 - alpha) * p1 + alpha * p2
            
            children = np.vstack((c1, c2))
            
            # Mutation (Annealing)
            progress = gen / self.n_generations
            current_sigma = self.mut_rate_start - (self.mut_rate_start - self.mut_rate_end) * progress
            
            mutation_mask = np.random.rand(*children.shape) < 0.3 # 30% chance to mutate a gene
            noise = np.random.normal(0, 1.0, children.shape) # Base noise
            
            # Scale noise by bounds
            ranges = np.array([h-l for l, h in self.bounds])
            scaled_noise = noise * ranges * current_sigma
            
            children[mutation_mask] += scaled_noise[mutation_mask]
            
            # Combine
            new_pop = np.vstack((elite, children))
            
            # Ensure population size stays constant (trim execution overflow)
            population = new_pop[:self.pop_size]
            
            # Clip bounds
            for i, (low, high) in enumerate(self.bounds):
                population[:, i] = np.clip(population[:, i], low, high)

        return population[0], self.history
```

```python
class FastEvolutionaryOptimizer:
    def __init__(self, 
                 simulator_func: Callable,
                 target_data: np.ndarray,
                 sensors: List[str],
                 bounds: List[Tuple[float, float]],
                 pop_size: int = 100,
                 n_generations: int = 50,
                 n_components: int = 2,
                 n_workers: int = 4): 
        
        self.sim_func = simulator_func
        self.target = target_data
        self.sensors = sensors
        self.bounds = bounds
        self.pop_size = pop_size
        self.n_generations = n_generations
        self.n_comps = n_components
        self.n_workers = n_workers
        self.n_genes = len(bounds)
        self.T = target_data.shape[0]
        
        self.mut_rate_start = 0.3
        self.mut_rate_end = 0.01
        self.history = {"fitness_best": [], "best_genes": []}

    def safe_initialize(self):
        """
        Retries random generation until we have 'pop_size' VALID individuals.
        """
        print("Initializing Safe Population (this might take a moment)...")
        valid_pop = []
        
        # We try in batches to be efficient
        batch_size = self.pop_size * 2 
        
        with ProcessPoolExecutor(max_workers=self.n_workers) as executor:
            while len(valid_pop) < self.pop_size:
                # 1. Generate a batch of random candidates
                candidates = np.random.rand(batch_size, self.n_genes)
                for i, (low, high) in enumerate(self.bounds):
                    candidates[:, i] = candidates[:, i] * (high - low) + low
                
                # 2. Test them in parallel
                tasks = [(ind, self.n_comps, self.T, self.sim_func, self.sensors, self.target) 
                         for ind in candidates]
                
                results = list(executor.map(evaluate_worker, tasks))
                
                # 3. Keep only the ones that didn't return -1e9
                for i, fit in enumerate(results):
                    if fit > -1e6: # Arbitrary threshold for "Valid"
                        valid_pop.append(candidates[i])
                        if len(valid_pop) >= self.pop_size:
                            break
                
                print(f"  -> Found {len(valid_pop)}/{self.pop_size} valid individuals...")
        
        return np.array(valid_pop[:self.pop_size])

    def run(self):
        print(f"--- Fast GA Started on {self.n_workers} Cores ---")
        
        # USE SAFE INITIALIZATION HERE
        population = self.safe_initialize()
        
        for gen in tqdm(range(self.n_generations), desc="Evolving"):
            
            # --- EVALUATION ---
            tasks = [(ind, self.n_comps, self.T, self.sim_func, self.sensors, self.target) 
                     for ind in population]

            with ProcessPoolExecutor(max_workers=self.n_workers) as executor:
                results = list(executor.map(evaluate_worker, tasks))
            
            fitnesses = np.array(results)
            
            # --- LOGGING ---
            best_idx = np.argmax(fitnesses)
            best_fit = fitnesses[best_idx]
            
            # If best fitness is still garbage, warn the user
            if best_fit < -1e5:
                print(f"Warning: Gen {gen} has no valid individuals. Check your bounds!")
            
            self.history["fitness_best"].append(best_fit)
            self.history["best_genes"].append(population[best_idx].copy())
            
            # --- SELECTION & BREEDING ---
            # Elitism
            sorted_idx = np.argsort(fitnesses)[::-1]
            n_elite = max(2, int(self.pop_size * 0.05))
            elite = population[sorted_idx[:n_elite]]
            
            # Selection
            n_needed = self.pop_size - n_elite
            n_pairs = int(np.ceil(n_needed / 2))
            
            # Tournament
            cands = np.random.randint(0, len(population), (n_pairs * 2, 3))
            scores = fitnesses[cands]
            winners = cands[np.arange(len(scores)), np.argmax(scores, axis=1)]
            parents = population[winners]
            
            # Crossover
            p1 = parents[:n_pairs]
            p2 = parents[n_pairs:]
            alpha = np.random.rand(n_pairs, 1)
            c1 = alpha * p1 + (1 - alpha) * p2
            c2 = (1 - alpha) * p1 + alpha * p2
            children = np.vstack((c1, c2))
            
            # Mutation
            progress = gen / self.n_generations
            curr_sigma = self.mut_rate_start - (self.mut_rate_start - self.mut_rate_end) * progress
            mask = np.random.rand(*children.shape) < 0.3
            noise = np.random.normal(0, 1.0, children.shape)
            ranges = np.array([h-l for l, h in self.bounds])
            children[mask] += (noise * ranges * curr_sigma)[mask]
            
            # Combine & Clamp
            population = np.vstack((elite, children))[:self.pop_size]
            for i, (low, high) in enumerate(self.bounds):
                population[:, i] = np.clip(population[:, i], low, high)

        return population[0], self.history
```

```python
import numpy as np
import matplotlib.pyplot as plt
import time
import warnings
from concurrent.futures import ProcessPoolExecutor
from tqdm import tqdm
from typing import Callable, List, Tuple

# ==============================================================================
# PART 1: THE ROBUST PARALLEL WORKER
# ==============================================================================
def evaluate_worker_fast(args):
    """
    Independent worker function for parallel execution.
    Handles:
    1. Physics Constraints (catches crashes)
    2. Time Decimation (simulates only every Nth step for speed)
    3. Matrix Broadcasting logic
    """
    genes, n_comps, T, sim_func, sensors, target, step_size = args

    # Suppress "invalid value encountered in sqrt" warnings
    with np.errstate(invalid='ignore', divide='ignore', over='ignore'):
        try:
            # --- 1. DECODE GENES ---
            # Genes = [Init_1, Init_2, ..., Rate_1, Rate_2...]
            initials = genes[:n_comps]
            rates = genes[n_comps:]
            
            # SPEED UP: Generate indices for only every Nth step
            # e.g., if T=100 and step=10 -> indices [0, 10, 20... 90]
            t_indices = np.arange(0, T, step_size)
            
            # Trajectory Calculation: Health = Init - (Rate * Time)
            # Shapes: Init(2,1) - Rate(2,1) * Time(1, M) -> Result(2, M) -> Transpose(M, 2)
            health_traj = initials[:, np.newaxis] - (rates[:, np.newaxis] * t_indices)
            health_traj = health_traj.T 

            # --- 2. RUN SIMULATOR ---
            # We assume sim_func accepts a batch of health parameters
            res = list(sim_func(health_traj, S=sensors))
            sim_data = np.array(res) # Shape (M, n_sensors)

            # --- 3. PHYSICS CHECK ---
            # If simulator returns NaNs or Infinities, it's a crash -> Return Penalty
            if np.isnan(sim_data).any() or np.isinf(sim_data).any():
                return -1e9 

            # --- 4. CALCULATE FITNESS ---
            # Slice target data to match the skipped steps
            target_sliced = target[::step_size]
            
            # Normalized RMSE (MAPE-like) to handle different sensor scales
            diff = np.abs(sim_data - target_sliced)
            rel_error = diff / (np.abs(target_sliced) + 1e-8)
            
            return -np.mean(rel_error)

        except Exception:
            # Catch any other Python errors (index out of bounds, etc.)
            return -1e9

# ==============================================================================
# PART 2: THE ULTRA-FAST OPTIMIZER CLASS
# ==============================================================================
class UltraFastOptimizer:
    def __init__(self, 
                 simulator_func: Callable,
                 target_data: np.ndarray,
                 sensors: List[str],
                 bounds: List[Tuple[float, float]],
                 pop_size: int = 50,       
                 n_generations: int = 30,  
                 n_components: int = 2,
                 n_workers: int = 4,
                 step_skip: int = 10):     
        
        self.sim_func = simulator_func
        self.target = target_data
        self.sensors = sensors
        self.bounds = bounds
        self.pop_size = pop_size
        self.n_generations = n_generations
        self.n_comps = n_components
        self.n_workers = n_workers
        self.n_genes = len(bounds)
        self.T = target_data.shape[0]
        self.step_skip = step_skip 
        
        # Adaptive Mutation Parameters
        self.mut_rate_start = 0.3
        self.mut_rate_end = 0.01
        
        self.history = {"fitness_best": [], "best_genes": []}

    def seed_population(self):
        """
        initializes population around a 'Safe Point' (Health=1.0, Decay=0.0).
        This prevents the 'invalid value' loop by ensuring Gen 0 is valid.
        """
        print("-> Seeding population around Safe Point (1.0, 0.0)...")
        pop = np.zeros((self.pop_size, self.n_genes))
        
        # Center: Health=1.0, Rate=0.0
        center_point = np.array([1.0]*self.n_comps + [0.0]*self.n_comps)
        
        for i in range(self.pop_size):
            # Add small noise to create diversity
            noise = np.random.normal(0, 0.005, self.n_genes) 
            candidate = center_point + noise
            
            # Clamp to bounds
            for j, (low, high) in enumerate(self.bounds):
                candidate[j] = np.clip(candidate[j], low, high)
            
            pop[i] = candidate
        return pop

    def run(self):
        print(f"--- GA Started on {self.n_workers} Cores (Skip={self.step_skip}) ---")
        
        population = self.seed_population()
        
        for gen in tqdm(range(self.n_generations), desc="Evolving"):
            
            # 1. PARALLEL EVALUATION
            tasks = [(ind, self.n_comps, self.T, self.sim_func, self.sensors, self.target, self.step_skip) 
                     for ind in population]

            with ProcessPoolExecutor(max_workers=self.n_workers) as executor:
                results = list(executor.map(evaluate_worker_fast, tasks))
            
            fitnesses = np.array(results)
            
            # 2. LOGGING
            best_idx = np.argmax(fitnesses)
            
            # Check if entire population failed
            if fitnesses[best_idx] < -1e5:
                print(f"Warning: Gen {gen} failed completely. Check bounds/simulator.")
            
            self.history["fitness_best"].append(fitnesses[best_idx])
            self.history["best_genes"].append(population[best_idx].copy())
            
            # 3. SELECTION (Elitism + Tournament)
            sorted_idx = np.argsort(fitnesses)[::-1]
            n_elite = max(2, int(self.pop_size * 0.1)) # Keep top 10%
            elite = population[sorted_idx[:n_elite]]
            
            n_needed = self.pop_size - n_elite
            n_pairs = int(np.ceil(n_needed / 2))
            
            # Vectorized Tournament
            cands = np.random.randint(0, len(population), (n_pairs * 2, 3))
            scores = fitnesses[cands]
            winners = cands[np.arange(len(scores)), np.argmax(scores, axis=1)]
            parents = population[winners]
            
            # 4. CROSSOVER (BLX-Alpha)
            p1 = parents[:n_pairs]
            p2 = parents[n_pairs:]
            alpha = np.random.rand(n_pairs, 1)
            c1 = alpha * p1 + (1 - alpha) * p2
            c2 = (1 - alpha) * p1 + alpha * p2
            children = np.vstack((c1, c2))
            
            # 5. ADAPTIVE MUTATION
            progress = gen / self.n_generations
            curr_sigma = self.mut_rate_start - (self.mut_rate_start - self.mut_rate_end) * progress
            
            mask = np.random.rand(*children.shape) < 0.4 # 40% mutation chance
            noise = np.random.normal(0, 1.0, children.shape)
            ranges = np.array([h-l for l, h in self.bounds])
            
            # Apply scaled noise
            children[mask] += (noise * ranges * curr_sigma)[mask]
            
            # 6. ASSEMBLE & CLAMP
            population = np.vstack((elite, children))[:self.pop_size]
            for i, (low, high) in enumerate(self.bounds):
                population[:, i] = np.clip(population[:, i], low, high)

        return population[0], self.history

# ==============================================================================
# PART 3: VISUALIZATION TOOLS
# ==============================================================================
def plot_results(history, best_genes, target_data, sim_func, sensors, n_comps):
    """
    Plots Convergence, Sensor Matching, and Health Degradation.
    Re-simulates the full trajectory (no skipping) for high-res plots.
    """
    
    # 1. CONVERGENCE
    plt.figure(figsize=(10, 4))
    plt.plot(history['fitness_best'], label='Best Fitness', color='royalblue', linewidth=2)
    plt.title("Optimization Convergence")
    plt.xlabel("Generation")
    plt.ylabel("Fitness (Negative Error)")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.show()

    # 2. RECONSTRUCT TRAJECTORY (Full Resolution)
    T = target_data.shape[0]
    initials = best_genes[:n_comps]
    rates = best_genes[n_comps:]
    t_steps = np.arange(T)
    
    # Init(2,1) - Rate(2,1) * Time(1, T) -> (2, T) -> Transpose
    health_traj = initials[:, np.newaxis] - (rates[:, np.newaxis] * t_steps)
    health_traj = health_traj.T 
    
    try:
        res = list(sim_func(health_traj, S=sensors))
        predicted_data = np.array(res)
    except Exception as e:
        print(f"Error regenerating plot data: {e}")
        return

    # 3. SENSOR MATCHING
    n_sensors = len(sensors)
    fig, axes = plt.subplots(1, n_sensors, figsize=(6 * n_sensors, 5))
    if n_sensors == 1: axes = [axes]
    
    for i, sensor in enumerate(sensors):
        axes[i].plot(target_data[:, i], 'k-', alpha=0.6, linewidth=2, label='Measured')
        axes[i].plot(predicted_data[:, i], 'r--', linewidth=2, label='GA Estimated')
        axes[i].set_title(f"Sensor: {sensor}")
        axes[i].legend()
        axes[i].grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
    
    # 4. INFERRED HEALTH
    plt.figure(figsize=(10, 4))
    labels = ['Compressor', 'Turbine'] # Add more if needed
    colors = ['purple', 'orange']
    for i in range(n_comps):
        lbl = labels[i] if i < len(labels) else f"Comp {i}"
        plt.plot(health_traj[:, i], linewidth=2, label=lbl, color=colors[i%len(colors)])
    plt.title("Inferred Health Degradation")
    plt.xlabel("Time (Flight Cycles)")
    plt.ylabel("Efficiency ($\eta$)")
    plt.legend()
    plt.grid(True)
    plt.show()

# ==============================================================================
# PART 4: EXECUTION BLOCK
# ==============================================================================
if __name__ == "__main__":
    
    # --- A. CONFIGURATION ---
    # Define TIGHT bounds to avoid crashing the simulator
    # Bounds: [Comp_Init, Turb_Init, Comp_Rate, Turb_Rate]
    BOUNDS = [
        (0.96, 1.01),   # Compressor Initial Health
        (0.96, 1.01),   # Turbine Initial Health
        (0.0, 0.0003),  # Compressor Decay Rate (Positive value)
        (0.0, 0.0003)   # Turbine Decay Rate
    ]
    
    SENSORS = ['P3', 'T5'] # Add your actual sensors here
    
    # --- B. LOAD DATA ---
    # Replace this with your actual loading logic
    # Example: target_measurements = np.load("my_flight_data.npy")
    # For now, I will create DUMMY data so you can run this block immediately to test.
    print("Generating Dummy Data for Test...")
    T_sim = 100
    true_health = np.linspace([0.98, 0.99], [0.96, 0.97], T_sim) # Linear decay
    # Mock simulator call to generate target (using h_SCSF if available, else Mock)
    try:
        # Assuming h_SCSF is defined in your environment
        target_measurements = np.array(list(h_SCSF(true_health, S=SENSORS)))
    except NameError:
        print("Warning: h_SCSF not found. Using random noise as placeholder.")
        target_measurements = np.random.rand(T_sim, 2)
        # Define a mock h_SCSF just to make the code runnable for testing
        def h_SCSF(health, S): return health # Mock pass-through

    # --- C. RUN OPTIMIZER ---
    import os
    cores = os.cpu_count()
    
    optimizer = UltraFastOptimizer(
        simulator_func=h_SCSF,      # Your simulator function
        target_data=target_measurements,
        sensors=SENSORS,
        bounds=BOUNDS,
        pop_size=50,                # Small population for speed
        n_generations=30,           # Sufficient for trajectory fitting
        n_components=2,             # Compressor + Turbine
        n_workers=cores - 1,        # Use all cores minus one
        step_skip=10                # Simulate only every 10th flight
    )
    
    start_time = time.time()
    best_genes, history = optimizer.run()
    end_time = time.time()
    
    print(f"\nOptimization Completed in {end_time - start_time:.2f} seconds.")
    print(f"Optimal Initial Health: {best_genes[:2]}")
    print(f"Optimal Decay Rates:    {best_genes[2:]}")
    
    # --- D. VISUALIZE ---
    plot_results(history, best_genes, target_measurements, h_SCSF, SENSORS, 2)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import time
import warnings
from concurrent.futures import ProcessPoolExecutor
from tqdm import tqdm
from typing import Callable, List, Tuple

# ==============================================================================
# PART 1: THE ROBUST PARALLEL WORKER
# ==============================================================================
def evaluate_worker_fast(args):
    """
    Independent worker function for parallel execution.
    Handles:
    1. Physics Constraints (catches crashes)
    2. Time Decimation (simulates only every Nth step for speed)
    3. Matrix Broadcasting logic
    """
    genes, n_comps, T, sim_func, sensors, target, step_size = args

    # Suppress "invalid value encountered in sqrt" warnings
    with np.errstate(invalid='ignore', divide='ignore', over='ignore'):
        try:
            # --- 1. DECODE GENES ---
            # Genes = [Init_1, Init_2, ..., Rate_1, Rate_2...]
            initials = genes[:n_comps]
            rates = genes[n_comps:]
            
            # SPEED UP: Generate indices for only every Nth step
            # e.g., if T=100 and step=10 -> indices [0, 10, 20... 90]
            t_indices = np.arange(0, T, step_size)
            
            # Trajectory Calculation: Health = Init - (Rate * Time)
            # Shapes: Init(2,1) - Rate(2,1) * Time(1, M) -> Result(2, M) -> Transpose(M, 2)
            health_traj = initials[:, np.newaxis] - (rates[:, np.newaxis] * t_indices)
            health_traj = health_traj.T 

            # --- 2. RUN SIMULATOR ---
            # We assume sim_func accepts a batch of health parameters
            res = list(sim_func(health_traj, S=sensors))
            sim_data = np.array(res) # Shape (M, n_sensors)

            # --- 3. PHYSICS CHECK ---
            # If simulator returns NaNs or Infinities, it's a crash -> Return Penalty
            if np.isnan(sim_data).any() or np.isinf(sim_data).any():
                return -1e9 

            # --- 4. CALCULATE FITNESS ---
            # Slice target data to match the skipped steps
            target_sliced = target[::step_size]
            
            # Normalized RMSE (MAPE-like) to handle different sensor scales
            diff = np.abs(sim_data - target_sliced)
            rel_error = diff / (np.abs(target_sliced) + 1e-8)
            
            return -np.mean(rel_error)

        except Exception:
            # Catch any other Python errors (index out of bounds, etc.)
            return -1e9

# ==============================================================================
# PART 2: THE ULTRA-FAST OPTIMIZER CLASS
# ==============================================================================
class UltraFastOptimizer:
    def __init__(self, 
                 simulator_func: Callable,
                 target_data: np.ndarray,
                 sensors: List[str],
                 bounds: List[Tuple[float, float]],
                 pop_size: int = 50,       
                 n_generations: int = 30,  
                 n_components: int = 2,
                 n_workers: int = 4,
                 step_skip: int = 10):     
        
        self.sim_func = simulator_func
        self.target = target_data
        self.sensors = sensors
        self.bounds = bounds
        self.pop_size = pop_size
        self.n_generations = n_generations
        self.n_comps = n_components
        self.n_workers = n_workers
        self.n_genes = len(bounds)
        self.T = target_data.shape[0]
        self.step_skip = step_skip 
        
        # Adaptive Mutation Parameters
        self.mut_rate_start = 0.3
        self.mut_rate_end = 0.01
        
        self.history = {"fitness_best": [], "best_genes": []}

    def seed_population(self):
        """
        initializes population around a 'Safe Point' (Health=1.0, Decay=0.0).
        This prevents the 'invalid value' loop by ensuring Gen 0 is valid.
        """
        print("-> Seeding population around Safe Point (1.0, 0.0)...")
        pop = np.zeros((self.pop_size, self.n_genes))
        
        # Center: Health=1.0, Rate=0.0
        center_point = np.array([1.0]*self.n_comps + [0.0]*self.n_comps)
        
        for i in range(self.pop_size):
            # Add small noise to create diversity
            noise = np.random.normal(0, 0.005, self.n_genes) 
            candidate = center_point + noise
            
            # Clamp to bounds
            for j, (low, high) in enumerate(self.bounds):
                candidate[j] = np.clip(candidate[j], low, high)
            
            pop[i] = candidate
        return pop

    def run(self):
        print(f"--- GA Started on {self.n_workers} Cores (Skip={self.step_skip}) ---")
        
        population = self.seed_population()
        
        for gen in tqdm(range(self.n_generations), desc="Evolving"):
            
            # 1. PARALLEL EVALUATION
            tasks = [(ind, self.n_comps, self.T, self.sim_func, self.sensors, self.target, self.step_skip) 
                     for ind in population]

            with ProcessPoolExecutor(max_workers=self.n_workers) as executor:
                results = list(executor.map(evaluate_worker_fast, tasks))
            
            fitnesses = np.array(results)
            
            # 2. LOGGING
            best_idx = np.argmax(fitnesses)
            
            # Check if entire population failed
            if fitnesses[best_idx] < -1e5:
                print(f"Warning: Gen {gen} failed completely. Check bounds/simulator.")
            
            self.history["fitness_best"].append(fitnesses[best_idx])
            self.history["best_genes"].append(population[best_idx].copy())
            
            # 3. SELECTION (Elitism + Tournament)
            sorted_idx = np.argsort(fitnesses)[::-1]
            n_elite = max(2, int(self.pop_size * 0.1)) # Keep top 10%
            elite = population[sorted_idx[:n_elite]]
            
            n_needed = self.pop_size - n_elite
            n_pairs = int(np.ceil(n_needed / 2))
            
            # Vectorized Tournament
            cands = np.random.randint(0, len(population), (n_pairs * 2, 3))
            scores = fitnesses[cands]
            winners = cands[np.arange(len(scores)), np.argmax(scores, axis=1)]
            parents = population[winners]
            
            # 4. CROSSOVER (BLX-Alpha)
            p1 = parents[:n_pairs]
            p2 = parents[n_pairs:]
            alpha = np.random.rand(n_pairs, 1)
            c1 = alpha * p1 + (1 - alpha) * p2
            c2 = (1 - alpha) * p1 + alpha * p2
            children = np.vstack((c1, c2))
            
            # 5. ADAPTIVE MUTATION
            progress = gen / self.n_generations
            curr_sigma = self.mut_rate_start - (self.mut_rate_start - self.mut_rate_end) * progress
            
            mask = np.random.rand(*children.shape) < 0.4 # 40% mutation chance
            noise = np.random.normal(0, 1.0, children.shape)
            ranges = np.array([h-l for l, h in self.bounds])
            
            # Apply scaled noise
            children[mask] += (noise * ranges * curr_sigma)[mask]
            
            # 6. ASSEMBLE & CLAMP
            population = np.vstack((elite, children))[:self.pop_size]
            for i, (low, high) in enumerate(self.bounds):
                population[:, i] = np.clip(population[:, i], low, high)

        return population[0], self.history

# ==============================================================================
# PART 3: VISUALIZATION TOOLS
# ==============================================================================
def plot_results(history, best_genes, target_data, sim_func, sensors, n_comps):
    """
    Plots Convergence, Sensor Matching, and Health Degradation.
    Re-simulates the full trajectory (no skipping) for high-res plots.
    """
    
    # 1. CONVERGENCE
    plt.figure(figsize=(10, 4))
    plt.plot(history['fitness_best'], label='Best Fitness', color='royalblue', linewidth=2)
    plt.title("Optimization Convergence")
    plt.xlabel("Generation")
    plt.ylabel("Fitness (Negative Error)")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.show()

    # 2. RECONSTRUCT TRAJECTORY (Full Resolution)
    T = target_data.shape[0]
    initials = best_genes[:n_comps]
    rates = best_genes[n_comps:]
    t_steps = np.arange(T)
    
    # Init(2,1) - Rate(2,1) * Time(1, T) -> (2, T) -> Transpose
    health_traj = initials[:, np.newaxis] - (rates[:, np.newaxis] * t_steps)
    health_traj = health_traj.T 
    
    try:
        res = list(sim_func(health_traj, S=sensors))
        predicted_data = np.array(res)
    except Exception as e:
        print(f"Error regenerating plot data: {e}")
        return

    # 3. SENSOR MATCHING
    n_sensors = len(sensors)
    fig, axes = plt.subplots(1, n_sensors, figsize=(6 * n_sensors, 5))
    if n_sensors == 1: axes = [axes]
    
    for i, sensor in enumerate(sensors):
        axes[i].plot(target_data[:, i], 'k-', alpha=0.6, linewidth=2, label='Measured')
        axes[i].plot(predicted_data[:, i], 'r--', linewidth=2, label='GA Estimated')
        axes[i].set_title(f"Sensor: {sensor}")
        axes[i].legend()
        axes[i].grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
    
    # 4. INFERRED HEALTH
    plt.figure(figsize=(10, 4))
    labels = ['Compressor', 'Turbine'] # Add more if needed
    colors = ['purple', 'orange']
    for i in range(n_comps):
        lbl = labels[i] if i < len(labels) else f"Comp {i}"
        plt.plot(health_traj[:, i], linewidth=2, label=lbl, color=colors[i%len(colors)])
    plt.title("Inferred Health Degradation")
    plt.xlabel("Time (Flight Cycles)")
    plt.ylabel("Efficiency ($\eta$)")
    plt.legend()
    plt.grid(True)
    plt.show()

# ==============================================================================
# PART 4: EXECUTION BLOCK
# ==============================================================================
if __name__ == "__main__":
    
    # --- A. CONFIGURATION ---
    # Define TIGHT bounds to avoid crashing the simulator
    # Bounds: [Comp_Init, Turb_Init, Comp_Rate, Turb_Rate]
    BOUNDS = [
        (0.96, 1.01),   # Compressor Initial Health
        (0.96, 1.01),   # Turbine Initial Health
        (0.0, 0.0003),  # Compressor Decay Rate (Positive value)
        (0.0, 0.0003)   # Turbine Decay Rate
    ]
    
    SENSORS = ['P3', 'T5'] # Add your actual sensors here
    
    # --- B. LOAD DATA ---
    # Replace this with your actual loading logic
    # Example: target_measurements = np.load("my_flight_data.npy")
    # For now, I will create DUMMY data so you can run this block immediately to test.
    # print("Generating Dummy Data for Test...")
    # T_sim = 100
    # true_health = np.linspace([0.98, 0.99], [0.96, 0.97], T_sim) # Linear decay
    # # Mock simulator call to generate target (using h_SCSF if available, else Mock)
    # try:
    #     # Assuming h_SCSF is defined in your environment
    #     target_measurements = np.array(list(h_SCSF(true_health, S=SENSORS)))
    # except NameError:
    #     print("Warning: h_SCSF not found. Using random noise as placeholder.")
    #     target_measurements = np.random.rand(T_sim, 2)
    #     # Define a mock h_SCSF just to make the code runnable for testing
    #     def h_SCSF(health, S): return health # Mock pass-through

    # # --- C. RUN OPTIMIZER ---
# --- B. LOAD DATA (YOUR CODE HERE) ---
    print("Generating Trajectory from Simulator...")
    
    # 1. Generate Ground Truth Health
    # NOTE: I increased size to 100. A trajectory needs length to show a trend.
    TRAJ_SIZE = 10 
    true_traj = simulate_degradation_trajectory(speed='fast', waterwash='rare', size=TRAJ_SIZE, seed=0)
    
    # Ensure shape is (TRAJ_SIZE, 2) for 2 components
    true_traj = np.asarray(true_traj, dtype=float)[:, :2] 

    # 2. Generate Sensor Measurements (The Target)
    # The GA needs a Numpy Array, not a List
    measured_series_list = list(h_SCSF(true_traj, S=SENSORS))
    target_measurements = np.array(measured_series_list) 

    print(f"-> Target Data Shape: {target_measurements.shape}")

    
    import os
    cores = os.cpu_count()
    
    optimizer = UltraFastOptimizer(
        simulator_func=h_SCSF,      # Your simulator function
        target_data=target_measurements,
        sensors=SENSORS,
        bounds=BOUNDS,
        pop_size=150,                # Small population for speed
        n_generations=30,           # Sufficient for trajectory fitting
        n_components=2,             # Compressor + Turbine
        n_workers=cores - 1,        # Use all cores minus one
        step_skip=10                # Simulate only every 10th flight
    )
    
    start_time = time.time()
    best_genes, history = optimizer.run()
    end_time = time.time()
    
    print(f"\nOptimization Completed in {end_time - start_time:.2f} seconds.")
    print(f"Optimal Initial Health: {best_genes[:2]}")
    print(f"Optimal Decay Rates:    {best_genes[2:]}")
    
    # --- D. VISUALIZE ---
    plot_results(history, best_genes, target_measurements, h_SCSF, SENSORS, 2)
```
