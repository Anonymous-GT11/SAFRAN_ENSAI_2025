```python
import numpy as np
import matplotlib.pyplot as plt
import time
from typing import Callable, List, Tuple

# ==============================================================================
# 1. THE SAFE WORKER (Now with Input Clipping)
# ==============================================================================
def evaluate_snapshot(args):
    """
    Evaluates a batch of individuals for a SINGLE flight (t).
    Includes Strict Bounds Enforcement.
    """
    genes_batch, sim_func, sensors, target_snapshot, hard_bounds = args
    
    fitnesses = []
    
    # 1. STRICT SAFETY CLIP
    # Before we even talk to the simulator, force genes into the legal box.
    # This prevents "degrad_turb too low" errors.
    safe_genes = np.zeros_like(genes_batch)
    for i, (low, high) in enumerate(hard_bounds):
        safe_genes[:, i] = np.clip(genes_batch[:, i], low, high)

    # Suppress warnings
    with np.errstate(invalid='ignore', divide='ignore'):
        try:
            # 2. Run Simulator on SAFE genes
            res = list(sim_func(safe_genes, S=sensors))
            sim_data = np.array(res) 
            
            # 3. Calculate Errors
            for i in range(len(safe_genes)):
                obs = sim_data[i]
                
                # Check for output NaNs (Physics failure)
                if np.isnan(obs).any() or np.isinf(obs).any():
                    fitnesses.append(-1e9)
                    continue
                
                # Normalized Error
                diff = np.abs(obs - target_snapshot)
                rel_error = np.mean(diff / (np.abs(target_snapshot) + 1e-8))
                fitnesses.append(-rel_error)
                
        except ValueError:
            # Catch specific simulator constraints
            return [-1e9] * len(genes_batch)
        except Exception as e:
            # Catch unexpected crashes
            # print(f"Crash: {e}") 
            return [-1e9] * len(genes_batch)
            
    return np.array(fitnesses)

# ==============================================================================
# 2. THE SEQUENTIAL TRACKER CLASS
# ==============================================================================
class SequentialGATracker:
    def __init__(self, 
                 simulator_func: Callable,
                 sensors: List[str],
                 global_bounds: List[Tuple[float, float]],
                 alpha: float = 0.005,   
                 pop_size: int = 40,    
                 n_generations: int = 10): 
        
        self.sim_func = simulator_func
        self.sensors = sensors
        self.global_bounds = global_bounds
        self.alpha = alpha
        self.pop_size = pop_size
        self.n_generations = n_generations
        self.n_genes = len(global_bounds)
        
        self.history = {
            "estimated_health": [],
            "estimated_sensors": [],
            "fitness_log": []
        }

    def get_dynamic_bounds(self, prev_solution: np.ndarray) -> List[Tuple[float, float]]:
        """
        Creates a tight search box around previous solution, 
        BUT clips strictly to the Global Hard Bounds.
        """
        new_bounds = []
        for i, (g_min, g_max) in enumerate(self.global_bounds):
            center = prev_solution[i]
            
            local_min = center - self.alpha
            local_max = center + self.alpha
            
            # Intersection of Local Box and Global Hard Constraints
            final_min = max(g_min, local_min)
            final_max = min(g_max, local_max)
            
            new_bounds.append((final_min, final_max))
        return new_bounds

    def run_trajectory(self, full_measurement_series):
        T = len(full_measurement_series)
        # Initialize middle of bounds if no prior info
        current_solution = np.array([np.mean(b) for b in self.global_bounds])
        
        print(f"--- Starting Sequential Tracking ({T} Flights) ---")
        start_time = time.time()
        
        for t in range(T):
            target = full_measurement_series[t]
            
            # A. Determine Bounds
            if t == 0:
                current_bounds = self.global_bounds
                generations_now = self.n_generations * 5 
            else:
                current_bounds = self.get_dynamic_bounds(current_solution)
                generations_now = self.n_generations

            # B. Initialize Population
            pop = np.random.rand(self.pop_size, self.n_genes)
            for i, (low, high) in enumerate(current_bounds):
                pop[:, i] = pop[:, i] * (high - low) + low
            
            best_step_ind = pop[0]
            best_step_fit = -1e9
            
            # C. Run Mini-GA
            for gen in range(generations_now):
                # Pass global bounds to worker for safety clipping
                fits = evaluate_snapshot((pop, self.sim_func, self.sensors, target, self.global_bounds))
                
                # Track best
                idx = np.argmax(fits)
                if fits[idx] > best_step_fit:
                    best_step_fit = fits[idx]
                    best_step_ind = pop[idx].copy()
                
                # Selection & Mutation
                n_elite = max(2, int(self.pop_size * 0.2))
                sorted_indices = np.argsort(fits)[::-1]
                elite = pop[sorted_indices[:n_elite]]
                
                offspring = []
                while len(offspring) < (self.pop_size - n_elite):
                    parent = elite[np.random.randint(0, n_elite)]
                    # Adaptive noise based on current bound width
                    noise_scale = [(h-l)*0.2 for l,h in current_bounds]
                    child = parent + np.random.normal(0, noise_scale)
                    offspring.append(child)
                
                pop = np.vstack((elite, offspring))
                
                # Clamp to current search bounds
                for i, (low, high) in enumerate(current_bounds):
                    pop[:, i] = np.clip(pop[:, i], low, high)

            # D. Update State
            current_solution = best_step_ind
            
            # E. Safe Logging
            self.history["estimated_health"].append(current_solution)
            self.history["fitness_log"].append(best_step_fit)
            
            # Estimate sensors (Protected Call)
            try:
                # Force clip before final check
                safe_sol = np.zeros_like(current_solution)
                for i, (low, high) in enumerate(self.global_bounds):
                    safe_sol[i] = np.clip(current_solution[i], low, high)
                    
                est_sens = list(self.sim_func([safe_sol], S=self.sensors))[0]
                self.history["estimated_sensors"].append(est_sens)
            except ValueError:
                # If it still fails, append NaNs or last known good
                self.history["estimated_sensors"].append(np.full(len(self.sensors), np.nan))
            
            if t % 10 == 0:
                print(f"Flight {t:3d}/{T} | Fit: {best_step_fit:.4f} | Health: {current_solution}")

        print(f"Tracking Completed in {time.time() - start_time:.2f}s")
        return self.history

# ==============================================================================
# 3. EXECUTION 
# ==============================================================================
if __name__ == "__main__":
    
    # 1. SETUP BOUNDS (MUST MATCH SIMULATOR HARD LIMITS)
    # The error said must be between 0.95 and 1.05.
    # We set it slightly tighter to be safe.
    GLOBAL_BOUNDS = [(0.9501, 1.0499), (0.9501, 1.0499)]
    
    SENSORS = ['P3', 'T5']
    
    # 2. GENERATE DUMMY DATA FOR TESTING
    # (Replace this with your actual loading code)
    print("Generating Test Data...")
    T_sim = 50
    # True health stays within legal limits
    true_traj = np.column_stack([
        np.linspace(1.0, 0.96, T_sim), # Comp
        np.linspace(1.0, 0.96, T_sim)  # Turb
    ])
    try:
        target_measurements = np.array(list(h_SCSF(true_traj, S=SENSORS)))
    except NameError:
        # Fallback if h_SCSF is not defined in this context
        target_measurements = np.random.rand(T_sim, 2)
        def h_SCSF(h, S): return h # Mock
    
    # 3. RUN TRACKER
    tracker = SequentialGATracker(
        simulator_func=h_SCSF,
        sensors=SENSORS,
        global_bounds=GLOBAL_BOUNDS, # <--- Passing the safe bounds
        alpha=0.01,  
        pop_size=30, 
        n_generations=5 
    )
    
    results = tracker.run_trajectory(target_measurements)
    
    # 4. PLOT
    est_health = np.array(results["estimated_health"])
    est_sensors = np.array(results["estimated_sensors"])
    
    fig, axs = plt.subplots(1, 2, figsize=(12, 5))
    
    # Sensors
    axs[0].plot(target_measurements[:, 0], 'k-', alpha=0.6, label='Measured P3')
    axs[0].plot(est_sensors[:, 0], 'r--', label='Tracked P3')
    axs[0].set_title("Sensor Tracking")
    axs[0].legend()
    
    # Health
    axs[1].plot(true_traj[:, 0], 'g-', alpha=0.6, label='True Health')
    axs[1].plot(est_health[:, 0], 'r--', label='Est Health')
    axs[1].set_title("Health Estimation")
    axs[1].legend()
    
    plt.tight_layout()
    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import time
from typing import Callable, List, Tuple

# ==============================================================================
# 1. VECTORIZED GA ENGINE
# ==============================================================================
class VectorizedSequentialTracker:
    def __init__(self, 
                 simulator_func: Callable,
                 sensors: List[str],
                 global_bounds: List[Tuple[float, float]],
                 alpha: float = 0.01,    # Trust region radius
                 pop_size: int = 50,     
                 n_generations: int = 10): 
        
        self.sim_func = simulator_func
        self.sensors = sensors
        self.global_bounds = global_bounds
        self.alpha = alpha
        self.pop_size = pop_size
        self.n_generations = n_generations
        self.n_genes = len(global_bounds)
        
        # Pre-compute numpy arrays for bounds for fast clipping
        self.global_min = np.array([b[0] for b in global_bounds])
        self.global_max = np.array([b[1] for b in global_bounds])
        
        self.history = {
            "estimated_health": [],
            "fitness_log": []
        }

    def evaluate_batch(self, population, target_snapshot):
        """
        Evaluates the ENTIRE population in ONE simulator call (Vectorized).
        """
        # 1. Run Simulator on the whole matrix (Pop_Size, N_Genes)
        # We assume h_SCSF treats input (N, M) as N instances of M components
        try:
            # Output expected: (Pop_Size, N_Sensors)
            # converting list to array handles the formatting
            sim_output = np.array(list(self.sim_func(population, S=self.sensors)))
            
            # 2. Vectorized Error Calculation
            # Check for NaNs (Physics failures) across rows
            # valid_mask: True if row is good, False if NaN/Inf
            valid_mask = np.isfinite(sim_output).all(axis=1)
            
            # Calculate Absolute Error
            # Target broadcasts: (1, Sensors) vs (Pop, Sensors)
            abs_diff = np.abs(sim_output - target_snapshot)
            
            # Normalize error: |Sim - Obs| / |Obs|
            # Add epsilon to avoid divide-by-zero
            norm_diff = abs_diff / (np.abs(target_snapshot) + 1e-8)
            
            # Mean error across sensors (axis=1)
            mean_errors = np.mean(norm_diff, axis=1)
            
            # 3. Assign Fitness
            fitnesses = -mean_errors
            
            # Apply "Death Penalty" to invalid rows efficiently
            # We set their fitness to -1e9
            fitnesses[~valid_mask] = -1e9
            
            return fitnesses
            
        except Exception as e:
            # Fallback if simulator crashes completely (e.g. dimension mismatch)
            print(f"Batch Sim Failed: {e}")
            return np.full(self.pop_size, -1e9)

    def run_trajectory(self, full_measurement_series):
        """
        Main Loop: Sequential Tracking
        """
        T = len(full_measurement_series)
        
        # Initial guess: Middle of bounds
        current_sol = (self.global_min + self.global_max) / 2.0
        
        print(f"--- Starting Vectorized Tracking ({T} steps) ---")
        start_time = time.time()
        
        for t in range(T):
            target = full_measurement_series[t]
            
            # A. Dynamic Bounds (Vectorized)
            if t == 0:
                # First step: Search everywhere
                lower_bounds = self.global_min
                upper_bounds = self.global_max
                gens = self.n_generations * 5 # Work harder on first point
            else:
                # Subsequent steps: Trust Region around previous
                # Box = [Prev - Alpha, Prev + Alpha]
                lower_bounds = np.maximum(self.global_min, current_sol - self.alpha)
                upper_bounds = np.minimum(self.global_max, current_sol + self.alpha)
                gens = self.n_generations

            # B. Initialize Population (Vectorized Uniform)
            # Shape: (Pop_Size, N_Genes)
            # Random scaling: min + rand * (max - min)
            rand_matrix = np.random.rand(self.pop_size, self.n_genes)
            population = lower_bounds + rand_matrix * (upper_bounds - lower_bounds)
            
            # C. Evolution Loop
            for gen in range(gens):
                # 1. Safety Clip (Whole Matrix)
                population = np.clip(population, self.global_min, self.global_max)
                
                # 2. Evaluate (Batch)
                fitnesses = self.evaluate_batch(population, target)
                
                # 3. Selection (Elitism)
                # Sort indices descending
                sorted_idx = np.argsort(fitnesses)[::-1]
                
                # Keep top 20%
                n_elite = int(self.pop_size * 0.2)
                elite = population[sorted_idx[:n_elite]]
                best_fit_now = fitnesses[sorted_idx[0]]
                
                # 4. Vectorized Mutation
                # We need to fill (Pop_Size - Elite) spots
                n_children = self.pop_size - n_elite
                
                # Randomly select parents from elite (Vectorized choice)
                parent_indices = np.random.randint(0, n_elite, size=n_children)
                parents = elite[parent_indices]
                
                # Generate Gaussian Noise Matrix
                # Scale noise to current bound width to be adaptive
                bound_widths = upper_bounds - lower_bounds
                noise = np.random.normal(0, 0.1, (n_children, self.n_genes)) * bound_widths
                
                children = parents + noise
                
                # 5. Next Generation
                population = np.vstack((elite, children))
                
                # (Loop ends, repeat generation)

            # D. Update State (Best of final generation)
            # Re-clip one last time to be safe
            population = np.clip(population, self.global_min, self.global_max)
            final_fits = self.evaluate_batch(population, target)
            best_idx = np.argmax(final_fits)
            
            current_sol = population[best_idx].copy()
            
            # E. Logging
            self.history["estimated_health"].append(current_sol)
            self.history["fitness_log"].append(np.max(final_fits))
            
            if t % 10 == 0:
                print(f"Step {t:3d} | Fit: {np.max(final_fits):.5f} | Sol: {current_sol}")

        elapsed = time.time() - start_time
        print(f"Completed in {elapsed:.2f}s ({elapsed/T:.4f}s per step)")
        return self.history

# ==============================================================================
# 2. EXECUTION & PLOTTING BLOCK
# ==============================================================================
if __name__ == "__main__":
    
    # --- CONFIGURATION ---
    # Global Limits (Hardware limits of the engine)
    # Simulator likely crashes outside 0.95 - 1.05
    HARD_BOUNDS = [(0.9501, 1.0499), (0.9501, 1.0499)] 
    SENSORS = ['P3', 'T5']
    
    # --- DUMMY SIMULATOR & DATA ---
    # (Replace this block with your imports)
    try:
        # Check if h_SCSF exists in memory
        h_SCSF
    except NameError:
        print("Warning: h_SCSF not found. Defining Mock Simulator.")
        def h_SCSF(pop, S):
            # Mock: Just returns the input (Health) + some noise
            # Pop shape (N, 2) -> Output (N, 2)
            return pop 

    # Generate Synthetic Truth (Sine wave to test tracking)
    T_sim = 100
    x = np.linspace(0, 4*np.pi, T_sim)
    true_health = np.column_stack([
        1.0 + 0.03 * np.sin(x),       # Component 1
        1.0 + 0.03 * np.cos(x) * 0.5  # Component 2
    ])
    # Generate Target Measurements
    target_data = np.array(list(h_SCSF(true_health, S=SENSORS)))
    
    # --- RUN OPTIMIZER ---
    tracker = VectorizedSequentialTracker(
        simulator_func=h_SCSF,
        sensors=SENSORS,
        global_bounds=HARD_BOUNDS,
        alpha=0.01,         # Trust region: 1% change allowed per step
        pop_size=5,        # Process 50 individuals in one batch
        n_generations=5     # 5 steps of refinement per flight
    )
    
    results = tracker.run_trajectory(target_data)
    
    # --- VISUALIZATION ---
    est_health = np.array(results["estimated_health"])
    
    fig, ax = plt.subplots(figsize=(10, 5))
    ax.plot(true_health[:, 0], 'k-', alpha=0.3, linewidth=4, label='True Comp 1')
    ax.plot(est_health[:, 0], 'r--', linewidth=2, label='Tracked Comp 1')
    
    ax.plot(true_health[:, 1], 'b-', alpha=0.3, linewidth=4, label='True Comp 2')
    ax.plot(est_health[:, 1], 'cyan', linestyle='--', linewidth=2, label='Tracked Comp 2')
    
    ax.set_title(f"Vectorized Tracking Results (Alpha={tracker.alpha})")
    ax.set_xlabel("Flight Cycle")
    ax.set_ylabel("Health Parameter")
    ax.legend()
    ax.grid(True, alpha=0.3)
    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import time
from typing import Callable, List, Tuple

# ==============================================================================
# 1. VECTORIZED GA ENGINE
# ==============================================================================
class VectorizedSequentialTracker:
    def __init__(self, 
                 simulator_func: Callable,
                 sensors: List[str],
                 global_bounds: List[Tuple[float, float]],
                 alpha: float = 0.01,    # Trust region radius
                 pop_size: int = 50,     
                 n_generations: int = 10): 
        
        self.sim_func = simulator_func
        self.sensors = sensors
        self.global_bounds = global_bounds
        self.alpha = alpha
        self.pop_size = pop_size
        self.n_generations = n_generations
        self.n_genes = len(global_bounds)
        
        # Pre-compute numpy arrays for bounds for fast clipping
        self.global_min = np.array([b[0] for b in global_bounds])
        self.global_max = np.array([b[1] for b in global_bounds])
        
        self.history = {
            "estimated_health": [],
            "fitness_log": []
        }

    def evaluate_batch(self, population, target_snapshot):
        """
        Evaluates the ENTIRE population in ONE simulator call (Vectorized).
        """
        # 1. Run Simulator on the whole matrix (Pop_Size, N_Genes)
        # We assume h_SCSF treats input (N, M) as N instances of M components
        try:
            # Output expected: (Pop_Size, N_Sensors)
            # converting list to array handles the formatting
            sim_output = np.array(list(self.sim_func(population, S=self.sensors)))
            
            # 2. Vectorized Error Calculation
            # Check for NaNs (Physics failures) across rows
            # valid_mask: True if row is good, False if NaN/Inf
            valid_mask = np.isfinite(sim_output).all(axis=1)
            
            # Calculate Absolute Error
            # Target broadcasts: (1, Sensors) vs (Pop, Sensors)
            abs_diff = np.abs(sim_output - target_snapshot)
            
            # Normalize error: |Sim - Obs| / |Obs|
            # Add epsilon to avoid divide-by-zero
            norm_diff = abs_diff / (np.abs(target_snapshot) + 1e-8)
            
            # Mean error across sensors (axis=1)
            mean_errors = np.mean(norm_diff, axis=1)
            
            # 3. Assign Fitness
            fitnesses = -mean_errors
            
            # Apply "Death Penalty" to invalid rows efficiently
            # We set their fitness to -1e9
            fitnesses[~valid_mask] = -1e9
            
            return fitnesses
            
        except Exception as e:
            # Fallback if simulator crashes completely (e.g. dimension mismatch)
            print(f"Batch Sim Failed: {e}")
            return np.full(self.pop_size, -1e9)

    def run_trajectory(self, full_measurement_series):
        """
        Main Loop: Sequential Tracking
        """
        T = len(full_measurement_series)
        
        # Initial guess: Middle of bounds
        current_sol = (self.global_min + self.global_max) / 2.0
        
        print(f"--- Starting Vectorized Tracking ({T} steps) ---")
        start_time = time.time()
        
        for t in range(T):
            target = full_measurement_series[t]
            
            # A. Dynamic Bounds (Vectorized)
            if t == 0:
                # First step: Search everywhere
                lower_bounds = self.global_min
                upper_bounds = self.global_max
                gens = self.n_generations * 5 # Work harder on first point
            else:
                # Subsequent steps: Trust Region around previous
                # Box = [Prev - Alpha, Prev + Alpha]
                lower_bounds = np.maximum(self.global_min, current_sol - self.alpha)
                upper_bounds = np.minimum(self.global_max, current_sol + self.alpha)
                gens = self.n_generations

            # B. Initialize Population (Vectorized Uniform)
            # Shape: (Pop_Size, N_Genes)
            # Random scaling: min + rand * (max - min)
            rand_matrix = np.random.rand(self.pop_size, self.n_genes)
            population = lower_bounds + rand_matrix * (upper_bounds - lower_bounds)
            
            # C. Evolution Loop
            for gen in range(gens):
                # 1. Safety Clip (Whole Matrix)
                population = np.clip(population, self.global_min, self.global_max)
                
                # 2. Evaluate (Batch)
                fitnesses = self.evaluate_batch(population, target)
                
                # 3. Selection (Elitism)
                # Sort indices descending
                sorted_idx = np.argsort(fitnesses)[::-1]
                
                # Keep top 20%
                n_elite = int(self.pop_size * 0.2)
                elite = population[sorted_idx[:n_elite]]
                best_fit_now = fitnesses[sorted_idx[0]]
                
                # 4. Vectorized Mutation
                # We need to fill (Pop_Size - Elite) spots
                n_children = self.pop_size - n_elite
                
                # Randomly select parents from elite (Vectorized choice)
                parent_indices = np.random.randint(0, n_elite, size=n_children)
                parents = elite[parent_indices]
                
                # Generate Gaussian Noise Matrix
                # Scale noise to current bound width to be adaptive
                bound_widths = upper_bounds - lower_bounds
                noise = np.random.normal(0, 0.1, (n_children, self.n_genes)) * bound_widths
                
                children = parents + noise
                
                # 5. Next Generation
                population = np.vstack((elite, children))
                
                # (Loop ends, repeat generation)

            # D. Update State (Best of final generation)
            # Re-clip one last time to be safe
            population = np.clip(population, self.global_min, self.global_max)
            final_fits = self.evaluate_batch(population, target)
            best_idx = np.argmax(final_fits)
            
            current_sol = population[best_idx].copy()
            
            # E. Logging
            self.history["estimated_health"].append(current_sol)
            self.history["fitness_log"].append(np.max(final_fits))
            
            if t % 10 == 0:
                print(f"Step {t:3d} | Fit: {np.max(final_fits):.5f} | Sol: {current_sol}")

        elapsed = time.time() - start_time
        print(f"Completed in {elapsed:.2f}s ({elapsed/T:.4f}s per step)")
        return self.history

# ==============================================================================
# 2. EXECUTION & PLOTTING BLOCK
# ==============================================================================
# ==============================================================================
# EXECUTION & PLOTTING BLOCK (Updated)
# ==============================================================================
if __name__ == "__main__":
    
    # --- 1. CONFIGURATION ---
    # Global Limits: slightly inside 0.95-1.05 to prevent simulator crashes
    HARD_BOUNDS = [(0.9501, 1.0499), (0.9501, 1.0499)] 
    
    # --- 2. GENERATE YOUR DATA ---
    print("Generating Trajectory...")
    
    # NOTE: I changed size=2 to size=100 so you can actually see the tracking happen over time.
    # If you leave it at 2, the code will finish instantly with no plot to show.
    TRAJ_SIZE = 20
    
    # Call your degradation simulator
    true_traj = simulate_degradation_trajectory(speed='fast', waterwash='rare', size=TRAJ_SIZE, seed=0)
    
    # Keep only the first 2 components (Compressor, Turbine)
    true_traj = np.asarray(true_traj, dtype=float)[:, :2]
    
    sensors = ['P3', 'T5']

    # Get measurements for the whole trajectory
    # h_SCSF likely returns a list of outputs; we convert to Numpy array for the Tracker
    measured_series = list(h_SCSF(true_traj, S=sensors))
    target_data = np.array(measured_series)

    print(f"-> Data Loaded. Steps: {len(target_data)}, Sensors: {sensors}")

    # --- 3. RUN OPTIMIZER ---
    tracker = VectorizedSequentialTracker(
        simulator_func=h_SCSF,      # Your simulator function
        sensors=sensors,
        global_bounds=HARD_BOUNDS,  # Safe bounds
        alpha=0.01,                 # Trust region: Allow 1% change per flight
        pop_size=50,                # 50 individuals per batch
        n_generations=5             # 5 refinement steps per flight
    )
    
    results = tracker.run_trajectory(target_data)
    
    # --- 4. VISUALIZATION ---
    est_health = np.array(results["estimated_health"])
    
    fig, ax = plt.subplots(figsize=(10, 5))
    
    # Plot Component 1 (Compressor)
    ax.plot(true_traj[:, 0], 'k-', alpha=0.5, linewidth=3, label='True Compressor')
    ax.plot(est_health[:, 0], 'r--', linewidth=2, label='Estimated Compressor')
    
    # Plot Component 2 (Turbine)
    ax.plot(true_traj[:, 1], 'b-', alpha=0.5, linewidth=3, label='True Turbine')
    ax.plot(est_health[:, 1], 'cyan', linestyle='--', linewidth=2, label='Estimated Turbine')
    
    ax.set_title(f"Sequential Tracking Results (Alpha={tracker.alpha})")
    ax.set_xlabel("Flight Cycle")
    ax.set_ylabel("Health Parameter (Efficiency)")
    ax.legend()
    ax.grid(True, alpha=0.3)
    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import time
from typing import Callable, List, Tuple
import matplotlib.cm as cm

class ProbabilisticTracker:
    def __init__(self, simulator_func: Callable, sensors: List[str], global_bounds: List[Tuple[float, float]],
                 alpha: float = 0.01, pop_size: int = 100, n_generations: int = 10): 
        
        self.sim_func = simulator_func
        self.sensors = sensors
        self.global_bounds = global_bounds
        self.alpha = alpha
        self.pop_size = pop_size
        self.n_generations = n_generations
        self.n_genes = len(global_bounds)
        self.global_min = np.array([b[0] for b in global_bounds])
        self.global_max = np.array([b[1] for b in global_bounds])
        
        # LOGGING EVERYTHING
        self.history = {
            "est_mean": [],   # Best guess (Mean of Elite)
            "est_std": [],    # Uncertainty (Std of Elite)
            "diversity": [],  # Population Variance over time
            "fitness_best": [],
            "all_fitness_history": [] # Store all fitness values for cloud plot
        }

    def evaluate_batch(self, population, target_snapshot):
        # (Same vectorized logic as before)
        try:
            sim_output = np.array(list(self.sim_func(population, S=self.sensors)))
            valid_mask = np.isfinite(sim_output).all(axis=1)
            abs_diff = np.abs(sim_output - target_snapshot)
            norm_diff = abs_diff / (np.abs(target_snapshot) + 1e-8)
            mean_errors = np.mean(norm_diff, axis=1)
            fitnesses = -mean_errors
            fitnesses[~valid_mask] = -1e9
            return fitnesses
        except:
            return np.full(self.pop_size, -1e9)

    def run_trajectory(self, full_measurement_series):
        T = len(full_measurement_series)
        current_sol = (self.global_min + self.global_max) / 2.0
        
        print(f"--- Probabilistic Tracking ({T} steps) ---")
        
        for t in range(T):
            target = full_measurement_series[t]
            
            # 1. Bounds Logic
            if t == 0:
                lower = self.global_min
                upper = self.global_max
                gens = self.n_generations * 5
            else:
                # Stronger constraint: Force search to stay close to previous MEAN
                lower = np.maximum(self.global_min, current_sol - self.alpha)
                upper = np.minimum(self.global_max, current_sol + self.alpha)
                gens = self.n_generations

            # 2. Init Population
            rand_matrix = np.random.rand(self.pop_size, self.n_genes)
            population = lower + rand_matrix * (upper - lower)
            
            step_fitness_cloud = [] # To store fitness of ALL gens in this step
            
            # 3. Evolution
            for gen in range(gens):
                population = np.clip(population, self.global_min, self.global_max)
                fitnesses = self.evaluate_batch(population, target)
                
                # Store for Visualization
                step_fitness_cloud.extend(fitnesses[fitnesses > -1e5].tolist())

                # Standard Selection/Mutation...
                sorted_idx = np.argsort(fitnesses)[::-1]
                n_elite = int(self.pop_size * 0.25) # Keep top 25% (Probabilistic)
                elite = population[sorted_idx[:n_elite]]
                
                # Make children
                n_children = self.pop_size - n_elite
                parent_indices = np.random.randint(0, n_elite, size=n_children)
                parents = elite[parent_indices]
                bound_widths = upper - lower
                noise = np.random.normal(0, 0.3, (n_children, self.n_genes)) * bound_widths
                children = parents + noise
                population = np.vstack((elite, children))

            # 4. PROBABILISTIC ESTIMATION
            # Instead of taking the single best, we take statistics of the Elite
            # This captures the "Cloud of Solutions"
            population = np.clip(population, self.global_min, self.global_max)
            final_fits = self.evaluate_batch(population, target)
            
            # Get Top 20%
            sorted_idx = np.argsort(final_fits)[::-1]
            top_n = int(self.pop_size * 0.2)
            top_population = population[sorted_idx[:top_n]]
            
            # Mean and Std
            step_mean = np.mean(top_population, axis=0)
            step_std = np.std(top_population, axis=0)
            
            # Update current solution for next step physics
            current_sol = step_mean 
            
            # 5. LOGGING
            self.history["est_mean"].append(step_mean)
            self.history["est_std"].append(step_std)
            self.history["diversity"].append(np.mean(step_std)) # Average spread
            self.history["fitness_best"].append(final_fits[sorted_idx[0]])
            self.history["all_fitness_history"].append(step_fitness_cloud) # Save all dots
            
            if t % 10 == 0:
                print(f"Flight {t:3d} | StdDev: {np.mean(step_std):.5f} | (High Std = Ill-posed)")

        return self.history

# ==============================================================================
# 2. ADVANCED VISUALIZATION
# ==============================================================================
def plot_probabilistic_results(history, true_traj):
    
    means = np.array(history["est_mean"])
    stds = np.array(history["est_std"])
    T = len(means)
    x_axis = np.arange(T)
    
    # --- PLOT 1: Estimated Components with Uncertainty Bands ---
    n_comps = means.shape[1]
    fig, axes = plt.subplots(n_comps, 1, figsize=(10, 3*n_comps), sharex=True)
    if n_comps == 1: axes = [axes]
    
    colors = ['r', 'b', 'g']
    
    for i in range(n_comps):
        ax = axes[i]
        c = colors[i % 3]
        
        # Plot True
        ax.plot(x_axis, true_traj[:, i], 'k-', linewidth=2, label='True')
        
        # Plot Mean Estimate
        ax.plot(x_axis, means[:, i], color=c, linestyle='--', linewidth=2, label='Est Mean')
        
        # Plot Uncertainty (Shaded Region)
        # 3 unknowns with 2 sensors -> High uncertainty expected
        # ax.fill_between(x_axis, 
        #                 means[:, i] - 2*stds[:, i], 
        #                 means[:, i] + 2*stds[:, i], 
        #                 color=c, alpha=0.2, label='95% Conf (Solution Space)')
        
        ax.set_ylabel(f"Comp {i+1} Health")
        ax.legend(loc='upper right')
        ax.grid(True, alpha=0.3)
        
    axes[0].set_title("Health Estimation with Solution Uncertainty")
    plt.tight_layout()
    plt.show()

    # --- PLOT 2: METRICS (Fitness, Diversity, Cloud) ---
    fig, axs = plt.subplots(3, 1, figsize=(10, 10), sharex=True)
    
    # A. Convergence (Best Fitness)
    axs[0].plot(history["fitness_best"], color='blue', linewidth=2)
    axs[0].set_title("Convergence: Best Fitness per Flight")
    axs[0].set_ylabel("Fitness (-Error)")
    axs[0].grid(True)
    
    # B. Diversity (Std Dev)
    axs[1].plot(history["diversity"], color='orange', linewidth=2)
    axs[1].set_title("Population Diversity (Mean Std Dev)")
    axs[1].set_ylabel("Spread")
    axs[1].grid(True)
    
    # C. Cloud Plot (All Fitness Values)
    # Flatten the list of lists
    all_fits = history["all_fitness_history"]
    for t in range(T):
        fits = all_fits[t]
        # Downsample if too many points for plotting speed
        if len(fits) > 50: fits = np.random.choice(fits, 50)
        
        # Scatter plot for this time step
        axs[2].scatter(np.full(len(fits), t), fits, 
                       c='purple', s=3, alpha=0.3)
        
    axs[2].set_title("Fitness Cloud (Distribution of Solutions)")
    axs[2].set_xlabel("Flight Cycle")
    axs[2].set_ylabel("Fitness Value")
    axs[2].set_ylim(min(history["fitness_best"])*1.2, 0) # Focus on top area
    
    plt.tight_layout()
    plt.show()



def plot_probabilistic_results(history, true_traj):
    
    means = np.array(history["est_mean"])
    stds = np.array(history["est_std"])
    T = len(means)
    x_axis = np.arange(T)
    
    # --- PLOT 1: Estimated Components with Uncertainty Bands ---
    n_comps = means.shape[1]
    fig, axes = plt.subplots(n_comps, 1, figsize=(10, 3*n_comps), sharex=True)
    if n_comps == 1: axes = [axes]
    
    colors = ['r', 'b', 'g', 'm']
    
    for i in range(n_comps):
        ax = axes[i]
        c = colors[i % len(colors)]
        
        # Plot True
        if true_traj.shape[1] > i:
            ax.plot(x_axis, true_traj[:, i], 'k-', linewidth=2, label='True')
        
        # Plot Mean Estimate
        ax.plot(x_axis, means[:, i], color=c, linestyle='--', linewidth=2, label='Est Mean')
        
        # Plot Uncertainty (Shaded Region)
        ax.fill_between(x_axis, 
                        means[:, i] - 2*stds[:, i], 
                        means[:, i] + 2*stds[:, i], 
                        color=c, alpha=0.2, label='95% Conf')
        
        ax.set_ylabel(f"Comp {i+1} Health")
        ax.legend(loc='upper right')
        ax.grid(True, alpha=0.3)
        
    axes[0].set_title("Health Estimation with Solution Uncertainty")
    plt.tight_layout()
    plt.show()

    # --- PLOT 2: METRICS (SymLog Scale) ---
    fig, axs = plt.subplots(3, 1, figsize=(10, 10), sharex=True)
    
    # A. Convergence (Best Fitness)
    axs[0].plot(history["fitness_best"], color='royalblue', linewidth=2)
    axs[0].set_title("Convergence: Best Fitness per Flight")
    axs[0].set_ylabel("Fitness (-Error)")
    
    # Symlog allows plotting negative values on a log-like scale
    # linthresh determines the region around 0 that remains linear
    axs[0].set_yscale('symlog', linthresh=1e-6) 
    axs[0].grid(True, which='both', alpha=0.3)
    
    # B. Diversity (Std Dev)
    axs[1].plot(history["diversity"], color='orange', linewidth=2)
    axs[1].set_title("Population Diversity (Mean Std Dev)")
    axs[1].set_ylabel("Spread (Log Scale)")
    
    # Use Log scale here (Standard Deviation is always positive)
    axs[1].set_yscale('log') 
    axs[1].grid(True, which='both', alpha=0.3)
    
    # C. Cloud Plot (All Fitness Values)
    all_fits = history["all_fitness_history"]
    for t in range(T):
        fits = all_fits[t]
        if len(fits) > 50: fits = np.random.choice(fits, 50) # Downsample
        
        # Scatter plot
        axs[2].scatter(np.full(len(fits), t), fits, 
                       c='purple', s=5, alpha=0.3)
        
    axs[2].set_title("Fitness Cloud (Solution Distribution)")
    axs[2].set_xlabel("Flight Cycle")
    axs[2].set_ylabel("Fitness (SymLog)")
    
    # Symlog again for the cloud
    axs[2].set_yscale('symlog', linthresh=1e-6)
    axs[2].grid(True, which='both', alpha=0.3)
    
    plt.tight_layout()
    plt.show()

    
# ==============================================================================
# 3. EXECUTION
# ==============================================================================
if __name__ == "__main__":
    
    # CONFIG: 3 Components, 2 Sensors
    # HARD_BOUNDS = [(0.9501, 1.0499)] * 3 
    HARD_BOUNDS = [
        (0.9501, 1.0499),  # Component 1 (e.g. Compressor)
        (0.9501, 1.0499),  # Component 2 (e.g. Turbine)
        (0.98, 1.02)       # Component 3 (New strict band)
    ]
    SENSORS = ['P3', 'T5']
    
    # GENERATE DATA
    TRAJ_SIZE = 10
    true_traj_raw = simulate_degradation_trajectory(speed='fast', waterwash='rare', size=TRAJ_SIZE, seed=0)
    true_traj = np.asarray(true_traj_raw, dtype=float)[:, :3] # Take 3 components
    
    target_data = np.array(list(h_SCSF(true_traj, S=SENSORS)))
    
    # RUN PROBABILISTIC TRACKER
    tracker = ProbabilisticTracker(
        simulator_func=h_SCSF,
        sensors=SENSORS,
        global_bounds=HARD_BOUNDS,
        alpha=0.005,    # Tight Alpha to force smoothness despite non-uniqueness
        pop_size=100,   # Higher population to map the solution space
        n_generations=6
    )
    
    results = tracker.run_trajectory(target_data)
    
    plot_probabilistic_results(results, true_traj)
```
